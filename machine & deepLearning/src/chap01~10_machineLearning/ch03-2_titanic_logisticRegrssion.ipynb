# 타이타닉 탑승자 생존 여부 예측

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
print(titanic)
```

```python
import statsmodels.api as sm
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
encoder.fit(titanic['sex']) # titanic의 성별 데이터를 학습, 실행 encoder : 머신러닝, 딥러닝의 모든 데이터는 문자형을 인식하지 못함 -> 숫자형 데이터로 변환
sex = encoder.transform(titanic['sex'])  # 문자형을 숫자형 데이터로 변환
titanic['sex'] = sex   # 변환된 데이터를 다시 해당 컬럼에 저장
titanic
```

```python
# from_formula 통계학적인 결과 분석
model = sm.Logit.from_formula('survived ~ pclass + sex + age + fare + parch + sibsp',
                              data=titanic)  # 종속변수 ~ 독립변수
logit = model.fit()
logit.summary()
```

```python
print('OR: 오즈비')
print(np.exp(logit.params))
```
