### hihell -> ihello : hihello
- Hi Hello RNN

```python
import numpy as np
import tensorflow as tf
```

```python
idx2char = ['h', 'i', 'e', 'l', 'o']

num_classes = 5
input_dim = 5
sequence_length = 6  # hihell
learning_rate = 0.1
```

```python
# x_data = [[0, 1, 0, 2, 4, 4]] # hihell
y_data = [[1, 0, 2, 3, 3, 4]]   # ihello  정답 데이터

# onehot encoding으로 전처리
x_one_hot = np.array([[[1,0,0,0,0], # 'h' 첫번째 인덱스 위치
                       [0,1,0,0,0], # 'i' 두번째 인덱스 위치
                       [1,0,0,0,0], # 'h' 첫번째 인덱스 위치
                       [0,0,1,0,0], # 'e' 세번째 인덱스 위치
                       [0,0,0,1,0], # 'l' 네번째 인덱스 위치
                       [0,0,0,1,0]]],# 'l'
                     dtype=np.float32)
                     
y_one_hot = tf.keras.utils.to_categorical(y_data, num_classes=num_classes) # to_categorical() 정수값으로 담겨진 데이터를 자동으로 onehot encoding방식으로 펼쳐줌
```

```python
print(x_one_hot.shape) # 학습
print(y_one_hot.shape) # 정답 

# 1, 6, 5  : 입력 6, 5개의 feature 특징, 3차원
# 정답 데이터도 동일(onhot encoding방식)
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(sequence_length, input_dim)),  # 입력 레이어
    tf.keras.layers.RNN(
        cell=tf.keras.layers.LSTMCell(units=num_classes),  # 5번 반복
        return_sequences=True
    )
])
```

```python
# 완전 연결형 레이어(fully connected layer)를 추가적으로 붙여줌
# TimeDistributed() 시간의 순차적 데이터를 처리

model.add(tf.keras.layers.TimeDistributed( \
    tf.keras.layers.Dense(units=num_classes, activation='softmax')))
```

```python
# Adam 알고리즘 적용

model.compile(loss='categorical_crossentropy',
   optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
             metrics=['accuracy'])
```

```python
model.fit(x_one_hot, y_one_hot, epochs=50)
```

```python
predictions = model.predict(x_one_hot) # 알파벳 'o'가 예측 되어야 함
```

```python
# 알파벳 'o'가 예측됨
for i, prediction in enumerate(predictions):
    print(prediction)
    result_str = [idx2char[c] for c in np.argmax(prediction, axis=1)]
    print('\t예측 결과: ', ''.join(result_str))
```
