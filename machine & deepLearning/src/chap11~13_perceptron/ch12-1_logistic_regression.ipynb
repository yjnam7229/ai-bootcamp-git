### 로지스틱 회귀 - Classification
#### 분류모델: 입출력 데이터는 정수값 (0 / 1 encoding)

```python
import tensorflow as tf
import numpy as np
```

```python
x_data = np.array([
            [1, 2],  # 입력의 feature는 2개, 강의 1시간, 공부 2시간 =>  0 불합격
            [2, 3],
            [3, 1],
            [4, 3],
            [5, 3],
            [6, 2]
         ])
y_data = np.array([[0], [0],[0],[1],[1],[1]])  # 정답 데이터는 정수값(분류모델)
```

```python
# tensorflow(텐서플로우) 이용한 신경망 구성

model = tf.keras.Sequential()
```

```python
# 입력 레이어 추가 .Input()
model.add(tf.keras.Input(shape=(2,))) # 한번에 2개의 입력이 들어오는 모형을 만듬, [1, 2]

# 출력 레이어 추가
# units=1 출력데이터 1개
# activation='sigmoid' 분류모델 예측 알고리즘
model.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))  

## 입출력 레이어 -> 신경망 구성
```

```python
# 손실함수 loss='binary_crossentropy' <- sigmoid' 분류모델 예측 알고리즘
# optimizer='sgd' == optimizer=tf.keras.optimizers.SGD (혼돈 행렬)
model.compile(loss='binary_crossentropy', optimizer='sgd', metrics=['accuracy'])
```

```python
model.summary()
```

```python
# fit 메소드로 트레이닝
loss = model.fit(x_data, y_data, epochs=500) # epochs= 고정된 값이 아님, 오차의 값이 더이상 없을때 까지 반복하면 됨
```

```python
# Accuracy report
print('Accuracy:', loss.history['accuracy'][-1])
```

```python
y_predict = model.predict(np.array([[3, 3], [4, 1]]))
y_predict
```
