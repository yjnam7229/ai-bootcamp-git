```python
import tensorflow as tf
import numpy as np
```

```python
# open, high, volume, low, close
xy = np.array([[828.659973, 833.450012, 908100, 828.349976, 831.659973],
               [823.02002, 828.070007, 1828100, 821.655029, 828.070007],
               [819.929993, 824.400024, 1438100, 818.97998, 824.159973],
               [816, 820.958984, 1008100, 815.48999, 819.23999],
               [819.359985, 823, 1188100, 818.469971, 818.97998],
               [819, 823, 1198100, 816, 820.450012],
               [811.700012, 815.25, 1098100, 809.780029, 813.669983],
               [809.51001, 816.659973, 1398100, 804.539978, 809.559998]])

# 가중치가 다른 경우 전처리 필요함
```

```python
x_data = xy[:, 0:-1]  # 학습 데이터
y_data = xy[:, [-1]]  # 정답 데이터, 주어진 데이터의 마지막 컬럼 항목 => 종가 데이터를 예측
```

```python
model = tf.keras.Sequential()
```

```python
model.add(tf.keras.Input(shape=(x_data.shape[1],)))

model.add(tf.keras.layers.Dense(
    units=y_data.shape[1],
    activation='linear'
))
```

```python
model.compile(loss='mse', 
              optimizer=tf.keras.optimizers.SGD(learning_rate=1e-10))  # learning_rate 값을 아주 작게 사용
              # optimizer=tf.keras.optimizers.SGD(learning_rate=0.1))  # 01.의 간격이 너무 커서 loss : nan, inf 가 되어 발산 -> learning_rate을 줄여가면서 학습시켜야 함


model.summary()
```

```python
loss = model.fit(x_data, y_data, epochs=1000)

# loss: nan w값 자체를 찾지를 못하고 발산이 되어 버림

# loss : nan  
```
