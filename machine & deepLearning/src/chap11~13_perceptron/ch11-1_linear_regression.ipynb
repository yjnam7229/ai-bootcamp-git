# 신경망 구현 순서  

1. Sequential 모형 클래스 객체 생성.  
2. add 메소드로 레이어(layer) 추가.  
   - 입력단부터 순차적으로 추가한다.
   - 레이어는 출력(units) 뉴런 갯수를 첫번째 인수로 받는다.
   - 최초의 레이어는 Input(shape=( , ))로 입력 크기를 설정해야 한다.
   - activation 인수로 활성화 함수 설정한다.
3. compile 메소드로 모형 완성.
   - loss 인수로 비용(cost)함수 설정.
   - optimizer 인수로 최적화 알고리즘 설정
   - metrics 인수로 트레이닝 단계에서 기록할 성능 기준 설정
4. fit 메소드로 트레이닝
   - epochs로 에포크(epoch) 횟수 설정

```python
# 텐서플로우 설치
# !pip install tensorflow

import tensorflow as tf
```

```python
import numpy as np

# H(x) = wx + b : w = -1, b = 1
x_train = np.array([1., 2., 3., 4.])  # 입력(학습) 데이터
y_train = np.array([0., -1., -2, -3.]) # 정답 데이터
```

```python
# 1. Sequential 모형 클래스 객체 생성.
tf.model = tf.keras.Sequential()  # 신경망의 역할을 할 수 있는 객체 생성
```

```python
# 2. add 메소드로 레이어(layer) 추가. 
# 입력 레이어 추가
# 전달받을 입력은 1개 shape=(1,)
tf.model.add(tf.keras.Input(shape=(1,))) # 단순회귀분석:독립변수 하나인 경우, tupple 자료형인 경우 값을 하나만 저장할 경우 ,를 붙임

# 출력 레이어 추가
tf.model.add(tf.keras.layers.Dense(units=1, activation="linear")) # 선형회귀 알고리즘으로 신경망으로 구축 units 1 출력 1개, activation : 적용 알고리즘

#### 신경망 구성 
```

```python
# 3. compile 메소드로 모형 완성.
# w : 가중치, b: 절편값 찾는 학습

sgd = tf.keras.optimizers.SGD(learning_rate=0.1)   # 경사하강법 알고리즘 사용 : S Gradient Descent 알고리즘, Hyper 파라미터(learning_rate : 학습율)
tf.model.compile(loss="mse", optimizer=sgd) # compile 연결, mse(mean square error 함수)

# SGD : 경사하강법 알고리즘으로 0.1씩 이동(learning_rate=0.1) 하면서 잔차를 구해라
# 손실함수 : mse() 는 잔차(Residual)
# 회귀분석에서 종속변수와 적합값(예상값)의 차이
# 잔차는 (종속변수 - 적합값)으로 정의
```

```python
tf.model.summary()
```

```python
# 학습 전, 최초 설정된 w 값 조회
weights = tf.model.layers[0].get_weights()
print('학습 전:' + str(weights))
```

```python
# 4. fit 메소드로 트레이닝
tf.model.fit(x_train, y_train, epochs=200)  # epochs 반복횟수
```

```python
# 학습을 통해 구한 w와 b의 값
weights = tf.model.layers[0].get_weights()
print('학습 후:' + str(weights))
```
