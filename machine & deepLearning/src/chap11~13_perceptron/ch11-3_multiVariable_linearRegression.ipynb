### 다중 선형 회귀 분석
##### 시험 성적 예측: one input (x1, x2, x3)을 사용한 회귀 분석



```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 학습 데이터
x_data = np.array([ [73., 80., 75.],
                    [93., 88., 93.], 
                    [89., 91., 90.], 
                    [96., 98., 100.], 
                    [73., 66., 70.]
                ])

# 정답 데이터
y_data = np.array([ [152.],
                    [185.],
                    [180.],
                    [196.],
                    [142.]
                ])
```

```python
# tensorflow(텐서플로우) 이용한 신경망 구성

tf.model = tf.keras.Sequential()
```

```python
# 입력 레이어 추가 .Input()
tf.model.add(tf.keras.Input(shape=(3,)))  # shape(3,) 입력이 3개 [73., 80., 75.]

# 출력 레이어 추가
tf.model.add(tf.keras.layers.Dense(units=1, activation='linear'))  # activation='linear' 선형회귀 모델 예측
```

```python

tf.model.summary()
```

```python
# 학습 알고리즘 적용
# 경사하강법 알고리즘 적용 tf.keras.optimizers.SGD
# Hyper 파라미터 : learning_rate=1e-5 
# sgd = tf.keras.optimizers.SGD(learning_rate=1e-5)

tf.model.compile(loss='mse', optimizer=tf.keras.optimizers.SGD(learning_rate=1e-5)) # 손실함수(비용함수) mse(mean square error)알고리즘 적용, 1e-5 : 1*10^{-5} (0.00001)를 의미하는 지수 표기법
tf.model.summary()
```

```python
# fit 메소드로 트레이닝
loss = tf.model.fit(x_data, y_data, epochs=500) # epochs 반복횟수
```

```python
# loss 값의 시각화

plt.plot(loss.history['loss'])
plt.title('Model loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(['Train'], loc='upper right')
plt.show()

# 오차값을 최소화하는 w값을 찾음
```

```python
# 학습을 통해 구한 w1, w2, w3, bias 값

tf.model.layers[0].get_weights()
```
