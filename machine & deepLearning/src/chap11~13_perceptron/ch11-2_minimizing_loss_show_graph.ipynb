```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
# H(x) = wx + b : w = -1, b = 1
x_train = np.array([1., 2., 3., 4.])
y_train = np.array([0., -1., -2., -3.])
```

```python
# 신경망 객체 Sequential 생성
tf.model = tf.keras.Sequential()
```

```python
# 입력 레이어 추가 .Input()
tf.model.add(tf.keras.Input(shape=(1,)))  # 전달받을 입력은 1개 shape=(1,)

# 출력 레이어 추가
# units=1 : 출력 데이터 1개
# activation='linear' 예측 알고리즘 선형회귀
tf.model.add(tf.keras.layers.Dense(units=1, activation='linear')) 

## 입출력 레이어 추가 -> 신경망 구성
## 입력, 출력 레이어 사이에는 수많은 레이어를 추가 할 수 있다 -> 딥러닝
```

```python
tf.model.summary()
```

```python
# 경사하강법 알고리즘 적용 tf.keras.optimizers.SGD
# compile 메소드로 모형 완성
sgd = tf.keras.optimizers.SGD(learning_rate=0.1) # 최소가 되는 rate를 찾아간다, 0.1의 간격으로 이동

tf.model.compile(loss='mse', optimizer=sgd)  # mse(mean squre error) : 오차의 제곱 평균을 구하는 손실 함수 적용, 접선의 기울기 양수 : 왼쪽으로 이동, 음수이면 오른쪽으로 이동 
```

```python
# fit 메소드로 트레이닝
loss = tf.model.fit(x_train, y_train, epochs=200)  # epochs 반복횟수
```

```python
# loss 값의 시각화

plt.plot(loss.history['loss'])
plt.title('Model loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(['Train'], loc='upper right')
plt.show()
```
