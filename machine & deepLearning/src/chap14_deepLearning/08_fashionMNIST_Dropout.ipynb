#### Xavier 정규분포 초기값 설정 + 드롭아웃 알고리즘

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# 정규화
train_scaled = train_input / 255.0
test_scaled = test_input / 255.0

train_scaled = train_scaled.reshape(-1, 28 * 28)
```

```python
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(train_scaled, train_target, \
                     test_size=0.2, random_state=42)
```

```python
model = tf.keras.Sequential()

# Xavier 정규분포 초기값 설정기
model.add(tf.keras.Input(shape=(28 * 28,)))

model.add(tf.keras.layers.Dense(
    units=1024,
    activation='relu',
    kernel_initializer='glorot_normal'  # Xavier 초기값 알고리즘 적용, kernel_initializer='glorot_normal', w값을 초기화
))
# 드랍아웃
# 하나의 레이어를 구성한 다음에 레이어 추가
# Dropout(0.3) : 70%의 노드만 연결해서 학습진행 -> 예측 모델 생성
model.add(tf.keras.layers.Dropout(0.3))  

model.add(tf.keras.layers.Dense(
    units=1024, 
    activation='relu',
    kernel_initializer='glorot_normal'))  
model.add(tf.keras.layers.Dropout(0.3))

model.add(tf.keras.layers.Dense(
    units=512, 
    activation='relu',
    kernel_initializer='glorot_normal'))
model.add(tf.keras.layers.Dropout(0.3))

model.add(tf.keras.layers.Dense(
    units=10, 
    activation='softmax',
    kernel_initializer='glorot_normal'))
```

```python
model.summary()
```

```python
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(train_scaled, train_target, epochs=50)
```

```python
model.evaluate(val_scaled, val_target)  # 한번도 학습하지 않은 검증 데이터
```
