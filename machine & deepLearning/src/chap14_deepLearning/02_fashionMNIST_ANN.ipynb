```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# 정규화
train_scaled = train_input / 255.0 # / 255.0 -> 0, 1사이의 값으로 매칭되게 변환


# 입력데이터: 2 dim -> 1 dim 변환
train_scaled = train_scaled.reshape(-1, 28 * 28)
train_scaled.shape
```

```python
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(train_scaled, train_target, \
                     test_size=0.2, random_state=42)  # 훈련 데이터를 80% 학습 데이터, 20% 테스트 데이터
```

```python
print(train_scaled.shape, train_target.shape)
print(val_scaled.shape, val_target.shape)

# 80% : 20% 비율
```

```python
# 신경망 구성

model = tf.keras.Sequential()

# 입력 레이어 구성
model.add(tf.keras.Input(shape=(28*28,))) # 784 입력 shape

# 출력 레이어 구성
model.add(tf.keras.layers.Dense(
    units=10,       # 10개의 클래스  0 : 티셔츠, 1 : 바지, 2 : 스웨터, 3 : 드레스 ...
    activation='softmax'  # 클래스의 갯수가 2개 이상인 경우
))

# 경사하강법 알고리즘 적용
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='sgd',
              metrics=['accuracy'])
```

```python
model.fit(train_scaled, train_target, epochs=100)
```

```python
model.evaluate(val_scaled, val_target)
```
