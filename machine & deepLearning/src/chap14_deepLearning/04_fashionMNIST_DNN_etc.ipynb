```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# 정규화
train_scaled = train_input / 255.0
test_scaled = test_input / 255.0

# 입력데이터: 2 dim -> 1 dim 변환
train_scaled = train_scaled.reshape(-1, 28 * 28)
train_scaled.shape
```

### 신경망 구성 방법 2
#### hidden layer 부터 생성 -> 선호하는 코딩 방식

```python
dense1 = tf.keras.layers.Dense(
    units=50,
    activation='sigmoid',
    name='hidden1'
)

dense2 = tf.keras.layers.Dense(
    units=100,
    activation='sigmoid',
    name='hidden2'
)

output = tf.keras.layers.Dense(
    units=10,
    activation='softmax',
    name='output'
)
```

```python
# 신경망 구성
# 먼저 출력단까지 완성했으면 망 구성 추가를 list형태로 구성
# model = tf.keras.Sequential()
model = tf.keras.Sequential([tf.keras.Input(shape=(784, )),
                             dense1, 
                             dense2,
                             output])
```

```python
model.summary()

# 50 * 784 = 39,250
# 50 * 100 + 100 = 5,100
# 100 * 10 + 10 = 1,010
```

### 신경망 구성 방법 3
#### Sequential 객체안에서 dircet 하게 작성 -> 가독성은 떨어짐

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(784,)),

    tf.keras.layers.Dense(
        units=50,
        activation='sigmoid',
        name='hidden1'
    ),

    tf.keras.layers.Dense(
        units=100,
        activation='sigmoid',
        name='hidden2'
    ),

    tf.keras.layers.Dense(
        units=10,
        activation='softmax',
        name='output'
    )
])
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='sgd',
              metrics=['accuracy'])
```

```python
loss = model.fit(train_scaled, train_target, \
                 epochs=100, batch_size=100)
```

```python
test_scaled = test_scaled.reshape(-1, 28*28)

model.evaluate(test_scaled, test_target)
```
