### Affine변환 : 행렬의 내적 계산
#### Affine변환 활성함수

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# 정규화
train_scaled = train_input / 255.0   # / 255.0 -> 0, 1사이의 값으로 매칭되게 변환
test_scaled = test_input / 255.0
```

```python
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(train_scaled, train_target, \
                     test_size=0.2, random_state=42)  # test_size=0.2 20% 검증용 데이터
```

```python
model = tf.keras.Sequential()

# 입력 레이어
# train_scaled = train_scaled.reshape(-1, 28 * 28) : 2차원 -> 1차원, 784개
model.add(tf.keras.Input(shape=(28, 28)))  # 이미지 특징을 그대로 가져감

model.add(tf.keras.layers.Flatten()) # Flatten() 함수 : 객체의 레이어 추가, 입력으로 전달된 데이터 2차원 -> 1차원으로 변환

# 추가 출력 레이어(히든 레이어)
# units(뉴런)  
model.add(tf.keras.layers.Dense(units=1024, activation='relu'))  # Rectified Linear Function
model.add(tf.keras.layers.Dense(units=512, activation='relu'))

# 출력 레이어
model.add(tf.keras.layers.Dense(units=10, activation='softmax'))

# 1. 입력 레이어 : 28 * 28개
# 2. Flatten() 레이어 추가 : 1차원으로 변환 784개 
# 3. 1024개 레이어 추가 : 28 * 28 * 1024 + 1024(바이어스 갯수) = 803,840
# 4. 512개 레이어 추가 : 1024 * 512 + 512(바이어스 갯수) = 524,800
```

```python
model.summary()
```

```python
# compile 메소드로 모형 완성
model.compile(optimizer='sgd',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(train_scaled, train_target, epochs=50)
```

```python
model.evaluate(val_scaled, val_target)
```
