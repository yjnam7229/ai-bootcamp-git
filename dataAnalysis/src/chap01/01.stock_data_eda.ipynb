## **-EDA(Exploratory Data Analysis)**
- 데이터를 본격적으로 모델링하기 전에 데이터의 특성을 파악하는 과정
- 즉, 데이터를 시각화하고 요약 통계를 활용하여 데이터의 특성과 패턴을 이해하고, 이후의 분석이나 머신러닝 모델링 방향을 결정하기 위한 탐색 과정을 의미
- 일반적으로 다음과 같은 작업을 포함
    - 데이터의 구조 확인 (shape, info())
    - 기술 통계 확인 (describe())
    - 결측치 및 이상치 확인
    - 변수의 분포 확인 (히스토그램, 박스플롯 등)
    - 변수 간 관계 분석 (산점도, 상관계수 등)
    - 데이터 품질 및 전처리 방향 결정

```python
import yfinance as yf
import pandas as pd

df = yf.download("NVDA", period="1y", interval="1d", multi_level_index = False)
df.head()
```

```python
df.tail()
```

```python
# 1. 기본 정보 확인
print("Shape:", df.shape)
print("\n====== Info: ======")
df.info()
print("\n====== Describe: ======")
print(df.describe())
```

```python
# 2. 컬럼명과 데이터 타입
print("컬럼명:", df.columns.tolist())
print("데이터 타입:")
print(df.dtypes)
```

```python
# 3. 첫 5개, 마지막 5개 행
print("첫 5개 행:")
print(df.head())
print("\n마지막 5개 행:")
print(df.tail())
```

```python
# 4. 종가 기본 통계
print(f"종가 최대값: {df['Close'].max():.0f}")
print(f"종가 최소값: {df['Close'].min():.0f}")
print(f"종가 평균값: {df['Close'].mean():.0f}")
```

```python
# 5. 거래량 통계
print(f"거래량 평균: {df['Volume'].mean():,.0f}")
print(f"거래량 중앙값: {df['Volume'].median():,.0f}")
print(f"거래량 표준편차: {df['Volume'].std():,.0f}")
```

```python
# 6. 고가-저가 차이 평균
high_low_diff = df['High'] - df['Low']
print(f"고가-저가 차이 평균: {high_low_diff.mean(): .0f}")
```

```python
# 7. 특정 컬럼 선택
selected_df = df[['Close', 'Volume']].copy()
print(selected_df.head())
```

```python
# 8. 최근 10개 데이터
recent_10 = df.tail(10)
print(recent_10)
```

```python
# 9. 종가 상위 10% 이상 필터링
threshold = df['Close'].quantile(0.9)
high_price = df[df['Close'] >= threshold]
print(f"종가 {threshold:.7f} 이상인 날: {len(high_price)}일")
```

```python
# 10. 평균 거래량보다 많은 날
avg_volume = df['Volume'].mean()
high_volume_days = df[df['Volume'] > avg_volume]
print(f'평균 거래량({avg_volume:,.0f})보다 많았던 날: {len(high_volume_days)}일')
```

```python
# 11. 거래량 내림차순 정렬
volume_sorted = df.sort_values('Volume', ascending=False)
print(volume_sorted['Volume'].head())
```

```python
# 12. 일일 변동폭
df['Daily_Change'] = df['Close'] - df['Open']
print(df[['Open', 'Close', 'Daily_Change']].head())
```

```python
# 13. 변동률
df['Change_Rate'] = ((df['Close'] - df['Open']) / df['Open'] * 100).round(2)
print(df[['Open', 'Close', 'Change_Rate']].head())
```

```python
# 14. 날짜 정보 추출
df['Year'] = df.index.year
df['Month'] = df.index.month
df['Weekday'] = df.index.day_name()
print(df[['Year', 'Month', 'Weekday']].head())
```

```python
# 15. 월별 집계
monthly_stats = df.groupby('Month').agg({
    'Close': 'mean',
    'Volume': 'sum'
}).round(0)
print(monthly_stats)
```

```python
# 16. 연도별 최고가, 최저가
yearly_stats = df.groupby('Year').agg({
    'High': 'max',
    'Low': 'min'
})
print(yearly_stats)
```

```python
# 17. 연속 상승일 계산
df['Price_Up'] = (df['Close'] > df['Open']).astype(int)
df['Consecutive_Up'] = df['Price_Up'].groupby((df['Price_Up'] != df['Price_Up'].shift()).cumsum()).cumsum()
max_consecutive = df['Consecutive_Up'].max()
print(f"최대 연속 상승일: {max_consecutive}일")
```
