##### **파생변수 생성**
###### 주어진 데이터에서 유의미한 값을 도출할 때 추가적으로 선언해준 변수

```python
import pandas as pd
import numpy as np
```

```python
# 예시 데이터 생성
sales_data = pd.DataFrame({
        'product' : ['노트북', '마우스', '키보드', '모니터', '스피커'],
        'price' : [1200000, 25000, 80000, 350000, 120000],
        'quantity' : [10, 50, 30, 15, 25],
        'cost' : [800000, 15000, 50000, 200000, 80000]
})

print(sales_data)
```

```python
# 1) 파생변수 생성

# 총 매출액 계산 (단가 * 수량)
sales_data['total_sales'] = sales_data['price'] * sales_data['quantity']  # 컬럼 추가, 기존 테이블에 존재하지 않으면 자동 추가

# 총 이익 계산 (현재매출 - 원가)
sales_data['profit'] = (sales_data['price'] - sales_data['cost']) * sales_data['quantity']

# 이익률 계산
sales_data['profit_rate'] = (sales_data['profit'] / sales_data['total_sales']) * 100
print(sales_data)  
```

```python
# 2) assign 메서드
# DataFrame 2차원 객체, 새로운 열을 추가하거나 기존 열을 수정한 새로운 dataFrame을 반환
sales_with_rank = sales_data.assign(
        sales_rank = sales_data['total_sales'].rank(ascending=False)  # 내림차순으로 ranking을 전달
)  

print(sales_with_rank)
# sales_data['profit'] = (sales_data['price'] - sales_data['cost']) * sales_data['quantity']
```

##### - **apply() 함수**

```python
# 제품 등급 분류 함수 정의

def classity_product_grage(row):
    if row['profit_rate'] >= 40:
        return 'A급'
    elif row['profit_rate'] >= 35:
        return 'B급'
    else:
        return 'C급'

# 실행 하면  함수에 대한 등록
```

```python
# 1) DataFrame에 함수 적용

# apply()를 사용하여 제품 등급 생성
sales_data['grade'] = sales_data.apply(classity_product_grage, axis=1) # 함수가 전달되면서 axis의 값을 1(y축 기준) profit_rate의 axis=1 이므로 profit_rate를 열을 기준으로 처리
print(sales_data)
```

```python
# 2) 람다 함수 사용

# 람다 함수를 사용한 가격 구간 분류
sales_data['price_category'] = sales_data['price'].apply(
        lambda x : '고가' if x >= 300000 else '중가' if x >= 100000 else '저가'  # 기능이 간단하면 rambda키워드 이용  rambda x : 기능 정의  [x 한개의 입력 매개변수 선언 : 기능 정의]
)

print(sales_data)
```

```python
# 3) 컬럼별 통계 계산(axis=0)

coluns_state = sales_data[['price', 'quantity', 'profit']].apply(np.mean) # 별도의 지정이 없으면 default값 axis=0
print(f"평균값\n {coluns_state}")
```

##### **- map() & applymap() 함수**

```python
print(sales_data)
```

```python
# 1) map - 딕셔너리 : map을 통한 dictionary 데이터 전달
# 카테고리 매핑 딕셔너리 생성

category_mapping = {
        '노트북': '컴퓨터',
        '마우스': '주변기기',
        '키보드': '주변기기',
        '모니터': '디스플레이',
        '스피커': '오디오'
}

grade_points = {'A급' : 100, 'B급' : 80, 'C급' : 60}

# map을 사용한 카테고리 변환
sales_data['product_category'] = sales_data['product'].map(category_mapping) # 입력으로 dictionary, 함수등이 전달될 수 있다

# 등급별 포인트 매핑
sales_data['grade_points'] = sales_data['grade'].map(grade_points)

print(sales_data[['product', 'product_category', 'grade', 'grade_points']])

```

```python
# 2) map - 함수
# 숫자 데이터에 함수 적용 (소수점 반올림)

numeric_columns = ['profit_rate']
sales_data['profit_rate_rounded'] = sales_data['profit_rate'].map(lambda x: round(x, 1))

print(sales_data[['product', 'profit_rate', 'profit_rate_rounded']])
```

##### **조건부 파생변수 생성 (np.where, np.select)**

```python
# 1) np.where (3개 parameter[boolen값, True, False]) : 3항 연산자
# np.where를 사용한 이진 분류

sales_data['is_profitable'] = np.where(sales_data['profit'] > 1000000, '고수익', '일반수익')
print(sales_data[['product', 'profit', 'is_profitable']])
```

```python
# 2) np.select
# np select를 사용한 다중 조건 분류
conditions = [
    sales_data['total_sales'] >= 10000000,
    sales_data['total_sales'] >= 5000000,
    sales_data['total_sales'] >= 2000000
]

choices = ['대형거래', '중형거래', '소형거리']

sales_data['transaction_size'] = np.select(conditions, choices, default='미니거래')
print(sales_data[['product', 'total_sales',  'transaction_size']])
```

```python
# 3) pd.cut

# pd.cut을 사용한 구간 분할
price_bins = [0, 50000, 200000, float('inf')]
price_labels = ['저가', '중가', '고가']

sales_data['price_segment'] = pd.cut(sales_data['price'], bins=price_bins, labels=price_labels) # bins 구간의 기준값

print(sales_data[['product', 'price','price_segment']])
```

##### **집계 함수를 활용한 파생변수 생성**

```python
# 제품 카테고리별 평균 수익률 계산
# transform 출력의 폼을 최종적으로 만듬, 공통의 범주값에 동일한 값에 채워 feature
sales_data['category_avg_profit_rate'] = sales_data.groupby('product_category')['profit_rate'].transform('mean') # 있는 갯수만큼 결과를 채워줌
sales_data

# 제품 카테고리별 총 매출 대비 비중 계산
sales_data['category_total_sales'] = sales_data.groupby('product_category')['total_sales'].transform('sum') # 동일한 그룹의 갯수만큼 채워서 출력
sales_data['sales_share_in_category'] = (sales_data['total_sales'] / sales_data['category_total_sales']) * 100

print(sales_data[['product', 'product_category', 'profit_rate', 'category_avg_profit_rate', 'sales_share_in_category']])
```

```python
# 정렬 후 누적 합계 계산 (매출 순위별 누적 매출)

sales_sorted = sales_data.sort_values('total_sales', ascending=False)
sales_sorted['cumulative_sales'] = sales_sorted['total_sales'].cumsum()
sales_sorted['cumulative_sales_ratio'] = (sales_sorted['cumulative_sales'] / sales_sorted['total_sales']) * 100
print(sales_sorted[['product', 'total_sales', 'cumulative_sales', 'cumulative_sales_ratio']].round(1))
```
