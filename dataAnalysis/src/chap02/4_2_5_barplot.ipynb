# 막대그래프(barplot)
###### 수치를 막대의 길이로 표현해 시각화하는 방법
###### 값들의 차이가 미미하거나 표시할 막대의 수가 많은 경우, 그래프를 이해하기 어려운 문제 발생
###### 이러한 경우 선그래프를 활용하거나 막대의 색상을 조절하여 시각화하는 것이 좋음


```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("tips")
df.head()
```

```python
df['size'].unique()
```

## tips 데이터를 활용한 막대그래프 생성

```python
sns.barplot(x=df['size'], y=df['tip'])

# min(df[df['size'] == 5]['tip'])
```

### 신뢰구간 제거

```python
sns.barplot(x=df['size'], y=df['tip'], errorbar=None)
```

-----
<div class="alert alert-warning">
[Tip] 막대의 길이가 비슷한 경우
</div>

```python
sns.barplot(x=df['day'], y=df['tip'])
```

------

### countplot: 개수 세기 전용 막대그래프

```python
sns.countplot(x=df['day'])
# sns.countplot(x=df['day'], hue=df['day'])
# sns.countplot(x=df['size'], hue=df['day'])
```

```python
# 범주형 데이터 셋을 기준으로 수행했을 때 유의미한 결과를 도출
df['day'].value_counts()
```
