# 산점도(scatterplot)

```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("tips")
df.head(30)

# print()
# df_titanic = sns.load_dataset("titanic")
# df_titanic.head()
```

## (기본 산점도) 음식가격과 팁의 상관관계

```python
# scatterplot : x축 변수와 y축 변수의 관계를 확인할 때 사용하는 그래프
# 각 변수의 값을 x-y 2차원 평면에 점으로 표시
# 상관관계의 강도는 점들이 직선에 얼마나 가깝게 분포되어 있는지로 판단
# 점들이 가까울수록 강한 상관관계를 의미
# 양의 상관관계 : 점들이 오른쪽 위로 상승하는 형태
# 음의 상관관계 : 점들이 오른쪽 아래로 하강하는 형태
# 무상관 : 점들이 특별한 패턴없이 분산된 형태

sns.scatterplot(x=df['total_bill'], y=df['tip'])

# sns.scatterplot(x=df_titanic['class'], y=df_titanic['fare']) 
```

## 색상을 활용한 그룹별 분석

```python
sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['sex'])
```

## (버블차트) 크기를 활용한 3차원 정보 표현

```python
sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['sex'], size=df['size'])
```

## 다차원 시각화

```python
# 4개 변수를 동시에 표현
sns.scatterplot(x=df['total_bill'], y=df['tip'],                
                hue=df['sex'],     # 색상으로 성별 구분
                size=df['size'],   # 크기로 인원수 표현
                style=df['time'],  # 모양으로 시간대 구분
                alpha=1.0)         # 투명도 조절
```
