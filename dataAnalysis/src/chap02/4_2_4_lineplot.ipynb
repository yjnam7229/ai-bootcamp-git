# 선그래프(lineplot)
###### 우리 일상 생활에서 자주 접하는 그래프 중 하나
###### 일기예보에서 이번주 예상 기온 변화를 나타낸 그래프, 시간에 따른 주가 변화량을 나타내는 그래프 등 주로 시간의 변화에 따른 값의 변화를 나타낼 때 많이 사용
###### x축에 연도, 월, 일과 같은 시간을 사용한다면 시간에 따른 데이터 변화 추이를 쉽게 확인 할 수 있음
###### 그렇기 때문에 선그래프는 변수의 변화, 트렌드, 변화율 정보가 중요한 경우 사용

```python
import seaborn as sns
import pandas as pd
```

## 기본 선 그래프 생성

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 42]

sns.lineplot(x=x, y=y)
# sns.lineplot(x=df['total_bill'], y=df['tip'])  
```

## tips 데이터를 활용한 선그래프 실습

```python
df = sns.load_dataset("tips")

sns.lineplot(x=df['size'], y=df['tip'])
# sns.lineplot(x=df['total_bill'], y=df['tip'])
```

-----
<div class="alert alert-warning">
[Tip] 신뢰구간의 변경
</div>

```python
print("min:{}, max:{}".format(
    min(df[df['size'] == 5]['tip']),
    max(df[df['size'] == 5]['tip'])
))
```

```python
# ci 100 : 신뢰구간을 100%로 설정하여 시각화

sns.lineplot(x=df['size'], y=df['tip'], errorbar=('ci', 100))  # errorbar라는 튜플의 형태로 정의
# sns.lineplot(x=df['size'], y=df['tip'])
# sns.regplot(x=df['total_bill'], y=df['tip'], scatter=False, ci=None)
```

```python
sns.lineplot(x=df['size'], y=df['tip'], errorbar=None)
```

----------

```python
sns.lineplot(x=df['size'], y=df['tip'], hue=df['day'],  errorbar=None) # 색상으로 day 구분
```

-----
<div class="alert alert-warning">
[Tip] hue 속성은 범주형(category) 변수를 지정해주세요!
</div>

```python
# hue 속성에 여러 개의 값이 포함된 변수를 지정한 경우
# hue 속성의 값은 범주형 값으로 (다양한 데이터 값이 존재하는 경우 의도한 대로 출력이 되지 않는다)
# sns.lineplot(x=df['day'], y=df['total_bill'], hue=df['sex'],  errorbar=None)
sns.lineplot(x=df['day'], y=df['total_bill'], hue=df['tip'],  errorbar=None)
```
