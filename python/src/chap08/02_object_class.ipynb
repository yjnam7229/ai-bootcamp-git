```python
# isinstance() 함수

# 학생 클래스를 선언합니다.
class Student:
    # def __init__():  # 초기화 함수
    def study(self):
        print("공부를 합니다.")

# 선생님 클래스를 선언합니다.
class Teacher:
    def teach(self):
        print("학생을 가르칩니다.")

# 교실 내부의 객체 리스트를 생성합니다.
classroom = [
    Teacher(), Student(), Student(), Student(), Student()    # 1명의 Teacher, 4명의 Student
]

# 반복을 적용해서 적절한 함수를 호출하게 합니다.
for person in classroom:
    if isinstance(person, Student):
        person.study()
    elif isinstance(person, Teacher):
        person.teach()
```

```python
# 클래스 변수 : 클래스명으로 접근함, 인스턴스들이 공유할 목적으로 사용하는 변수, 생성자 함수 호출과 상관없이 클래스 이름만 만나면 접근이 가능
# self.이 붙으면 나 혼자 쓰는 인스턴스 변수, 클래스 밑에 그냥 적으면 모두가 공유하는 클래스 변수
# 클래스를 선언합니다.
class Student:
    count = 0  # 클래스 변수로 선언되어 있으면 먼저 메모리를 할당, 객체들이 값을 공유하면서 사용해야 할 데이터

    def __init__(self, name, korean, math, english, science): # 객체가 생성될 때마다 호출, 생성자 함수가 호출될때 실행
        # 인스턴스 변수 초기화
        self.name = name  # 해당 메모리에 접근하는 방법 self. 메모리에 할당된 시작주소 값, self는 첫번째 매개변수로 선언
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science
        
        # 클래스 변수 설정
        Student.count += 1
        print("{}번째 학생이 생성되었습니다.".format(Student.count))

# 학생 리스트를 선언합니다.
students = [
    Student("윤인성", 87, 98, 88, 95), # 생성자 함수 호출이 되면서 메모리 공간이 할당
    Student("연하진", 92, 98, 96, 98),
    Student("구지연", 76, 96, 94, 90),
    Student("나선주", 98, 92, 96, 92),
    Student("윤아린", 95, 98, 98, 98),
    Student("윤명월", 64, 88, 92, 92)
]

# 출력합니다.
print()
print("현재 생성된 총 학생 수는 {}명입니다.".format(Student.count))
```

```python
# 클래스 함수

# 클래스를 선언합니다.
class Student:
    # 클래스 변수 : 인스턴스들이 공유할 목적으로 선언해주는 변수
    count = 0
    students = []

    # 클래스 함수 
    # 클래스 함수 안에 데코레이터 정의
    @classmethod   
    def print(cls):  # self가 아니라 cls는 클래스의 약어(식별자)
        print("------------- 학생 목록 ----------------")
        print("이름\t 총점\t 평균")
        for student in cls.students:
            print(str(student))
        print("------------ ------------- ------------")

    # 인스턴스 함수 : 메모리가 할당되었다는 의미
    def __init__(self, name, korean, math, english, science):
        self.name = name
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science
        Student.count += 1
        Student.students.append(self)

    def get_sum(self):  # 인스턴스 함수
        return self.korean + self.math +\
            self.english + self.science

    def get_average(self):
        return self.get_sum() / 4

    def __str__(self):
        return "{}\t{}\t{}".format(\
            self.name,\
            self.get_sum(),\
            self.get_average())

# 학생 리스트를 선언합니다.
Student("윤인성", 87, 98, 88, 95)
Student("연하진", 92, 98, 96, 98)
Student("구지연", 76, 96, 94, 90)
Student("나선주", 98, 92, 96, 92)
Student("윤아린", 95, 98, 98, 98)
Student("윤명월", 64, 88, 92, 92)
Student("김미화", 82, 86, 98, 88)
Student("김연화", 88, 74, 78, 92)
Student("박아현", 97, 92, 88, 95)
Student("서준서", 45, 52, 72, 78)

# 현재 생성된 학생을 모두 출력합니다.
Student.print()
```

```python
# garbage collector-01

class Test:
    def __init__(self, name):
        self.name = name  # 할당된 메모리에 접근 self.
        print("{} - 생성되었습니다".format(self.name))
    def __del__(self):  # 콜백 함수
        print("{} - 파괴되었습니다".format(self.name))

Test("A") # 할당된 메모리가 없음 -> 메모리를 바라보는 변수가 없음(__del__ 함수를 파이썬이 호출하면서 재사용 가능한 메모리로 만듬)
Test("B")
Test("C")
```

```python
# private 변수와 getter/setter
# private 변수 __로 선언된 사용자가 선언한 함수

# 모듈을 가져옵니다.
import math

# 클래스를 선언합니다.
class Circle:
    def __init__(self, radius):
        self.__radius = radius    # 변수의 이름 정의할 때 __를 붙인다, private 하므로 외부에서 절대 접근 불가, 외부에서 선언된 circle.__radius = -2 와 영역이 다름

    def get_circumference(self):
        return 2 * math.pi * self.__radius
    
    def get_area(self):
        return  math.pi * self.__radius ** 2

# 원의 둘레와 넓이를 구합니다.
circle = Circle(10)

print("# 원의 둘레와 넓이를 구합니다.")
print("원의 둘레:", circle.get_circumference())
print("원의 넓이:", circle.get_area())
print()

# __radius에 접근합니다.
print("# __radius에 접근합니다.")
circle.__radius = -2  # __radius는 클래스 밖의 변수 : private 안에 있는 __radius 변수와 다름 , 외부에서 새롭게 사용할 수 있게끔 변수를 선언(새롭게 메모리 할당)
print(circle.__radius)  # error : private 변수 클래스의 밖에서는 접근 불가, 영역 내부에 있는 circle.__radius = -2 선언하면 에러 발생 안함
```

```python
# private 변수와 getter/setter
#  getter/setter

# 모듈을 가져옵니다.
import math

# 클래스를 선언합니다.
class Circle:
    def __init__(self, radius):
        self.__radius = radius    # 변수의 이름 정의할 때 __를 붙인다

    def get_circumference(self):
        return 2 * math.pi * self.__radius
    
    def get_area(self):
        return  math.pi * self.__radius ** 2

    # 게터와 세터를 선언합니다.
    def get_radius(self):  # 해당 변수의 값을 참조
        return self.__radius

    def set_radius(self, value):  # 해당 변수의 값을 변경
        self.__radius = value

# 원의 둘레와 넓이를 구합니다.
circle = Circle(10)
print("# 원의 둘레와 넓이를 구합니다.")
print("원의 둘레:", circle.get_circumference())
print("원의 넓이:", circle.get_area())
print()

# 간접적으로 __radius에 접근합니다.
print("# __radius에 접근합니다.")
print(circle.get_radius())
print()

# 원의 둘레와 넓이를 구합니다.
circle.set_radius(2)
print("# 반지름을 변경하고 원의 둘레와 넓이를 구합니다.")
print("원의 둘레:", circle.get_circumference())
print("원의 넓이:", circle.get_area())
```

```python
# 데코레이터를 사용한 게터와 세터

# 모듈을 가져옵니다.
import math

# 클래스를 선언합니다.
class Circle:
    def __init__(self, radius):
        self.__radius = radius  # __radius private 변수
    def get_circumference(self):
        return 2 * math.pi *  self.__radius
    def get_area(self):
        return math.pi * (self.__radius ** 2)

    # 게터와 세터를 선언합니다.
    # 파이썬에선 게터를 만들지 않으면 세터를 만들 수 없다.
    # 데코레이터
    @property
    def radius(self):  # getter 메서드
        return self.__radius  
    
    @radius.setter  # setter 메서드
    def radius(self, value):
        if value <= 0:
            raise TypeError("길이는 양의 숫자여야 합니다.")  # 강제 Exception
        self.__radius = value  # 변경

# 원의 둘레와 넓이를 구합니다.
print("# 데코레이터를 사용한 Getter와 Setter")
circle = Circle(10)
print("원래 원의 반지름: ", circle.radius)
circle.radius = 2
print("변경된 원의 반지름: ", circle.radius)
print()

# 강제로 예외를 발생시킵니다.
print("# 강제로 예외를 발생시킵니다.")
circle.radius = -10
```

```python
# 상속

# 부모 클래스를 선언합니다.
class Parent:
    def __init__(self):
        self.value = "테스트"
        print("Parent 클래스의 __init()__ 메서드가 호출되었습니다.")
    def test(self):
        print("Parent 클래스의 test() 메서드입니다.")

# 자식 클래스를 선언합니다.
class Child(Parent): # Parent 부모클래스 상속 : 자식클래스(부모클래스)
    def __init__(self):
        super().__init__() # 부모 클래스의 __init__() 함수를 호출(반드시 초기화 함수 정의)
        print("Child 클래스의 __init__() 함수가 호출되었습니다.")

class Child_2(Parent): # 자식클래스(부모클래스)
    pass

# 자식 클래스의 인스턴스를 생성하고 부모의 메서드를 호출합니다.
child = Child() # 인스턴스 생성되면서 생성자 함수 호출, 할당된 시작주소 값이 child 에 담김
child.test() # 자식의 인스턴스를 통해서 호출
print(child.value) # 부모클래스에 정의된 변수값에 접근 가능
```

```python
# 상속의 활용 - 예외 클래스 만들기

class CustomException(Exception):
    def __init__(self):
        super().__init__() # 부모에 정의되어 있는 초기화 함수를 호출 Exception을 상속하는 구조 
        
raise CustomException # raise Exception에 정의된 키워드 : 자료형의 예외를 강제로 발생, CustomException 이름으로 예외를 발생
```

```python
# 자식 클래스에 새로운 함수 정의하기

# 사용자 정의 예외를 생성합니다.
class CustomException(Exception):
    def __init__(self, message, value):
        super().__init__()
        self.message = message
        self.value = value

    def __str__(self):
        return self.message

    def print(self): # 자식 메서드에 추가적인 기능 구현
        print("###### 오류 정보 ######")
        print("메시지:", self.message)
        print("값:", self.value)

# 예외를 발생시켜 봅니다.
try:
    raise CustomException("딱히 이유 없음", 365) # 생성자 함수 호출, [message] [value]
except CustomException as ex:
    ex.print()  # CustomException 자료형안에 print()메서드 호출
```
