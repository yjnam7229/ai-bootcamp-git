## 4. 숫자와 문자열의 다양한 기능
### - 문자열의 format() 함수

```python
# format() 힘수로 숫자를 문자열로 변환하기
string_a = "{}".format(10)
format_a = "{}만원".format(5000)
format_b = "파이썬 열공하여 첫 연봉 {}만원 만들기".format(9000)
format_c = "{}{}{}".format(3000, 4000, 5000)
format_d = "{} {} {}".format(3.14, "문자열", True)

print(string_a)
print(type(string_a))

print(format_a)
print(format_b)
print(format_c)
print(format_d)
#print(str(6000) + "만원")
#print(type(format_a))
```

```python
# 정수
#output_a = "{:d}".format(52)
#output_a = "{:d}".format("홍길동")  에러
output_a = "{:s}".format("홍길동")

# 특정 칸에 출력하기
output_b = "{:5d}".format(52)  #양수
output_c = "{:10d}".format(-52)  #음수

# 빈칸을 0으로 채우기
output_d = "{:05d}".format(52) #양수
output_e = "{:05d}".format(-52) #음수

# 기호와 함께 출력하기
output_f = "{:+d}".format(52)
output_g = "{:+d}".format(-52)
output_h = "{: d}".format(52) # 양수: 기호 부분 공백

print(output_a)
print(output_b)
print(output_c)
print(output_d)
print(output_e)
print("# 기호와 함께 출력하기")
print(output_f)
print(output_g)
print(output_h)
```

```python
# 조합하기
output_h = "{:+5d}".format(52) # 기호를 뒤로 밀기: 양수
output_i = "{:+5d}".format(-52) # 기호를 뒤로 밀기: 음수
output_j = "{:=+5d}".format(52) # 기호를 앞으로 밀기: 양수
output_k = "{:=+5d}".format(-52) # 기호를 앞으로 밀기: 음수
output_l = "{:=+05d}".format(52) # 0으로 채우기: 양수
output_m = "{:=+05d}".format(-52) # 0으로 채우기: 음수

print("# 조합하기")
print(output_h)
print(output_i)
print(output_j)
print(output_k)
print(output_l)
print(output_m)
```

```python
output_a = "{:f}".format(52.273) # 소숫점 6째 자리로 기본셋팅 되어 있음
output_b = "{:15f}".format(52.273) # 15칸 만들기 15라는 숫자는 decimal
output_c = "{:+15f}".format(52.273) # 15칸에 부호 추가하기
output_d = "{:+015f}".format(52.273) # 15칸에 부호 추가하고 0으로 채우기

print(output_a)
print(output_b)
print(output_c)
print(output_d)
```

```python
output_a = 52.0
output_b = "{:g}".format(output_a) # :g 실수형 .0이면 실수 부분 생략하고 정수값만 출력(의미없는 소수점 제거)   :f 는 소숫점 6째자리까지 출력 
print(output_a)
print(output_b)
```

```python
a = "10 20 30 40 50".split(" ") # 공백전 까지 문자열, 실행결과는 list
print(a)

b = "10_20_30_40_50".split("_") # _ 전 까지 문자열, 실행결과는 list
print(b)

c = "10,20,30,40,50".split(",") # , 전 까지 문자열, 실행결과는 list
print(c)

d = "10,20,30,40,50".split("-") # , 전 까지 문자열, 실행결과는 list
print(d)

d = "10-20-30-40-50".split("-") # , - 이 없어서 하나의 문자열로 인식
print(d)
```
