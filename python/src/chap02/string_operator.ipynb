# 문자열 연산자

```python
print("안녕 " + "하세요")
print("안녕 " + "!")
#print("안녕 " + 10) # 문자열과 숫자의 조합은 지원 불가
#print("안녕 " / 10)
print("안녕 " * 10)
```

```python
# 문자 선택 연산자 [인덱스]
print("안녕하세요"[0])
print("안녕하세요"[1])
print("안녕하세요"[2])
```

```python
# 문자 선택 연산자 [음수 인덱스]
print("안녕하세요"[-1]) # 현재 문자열에서 제일 마지막 문자 선택
print("안녕하세요"[-2])
```

```python
# 문자열 범위 선택 연산자(슬라이싱) [:]
print("안녕하세요"[1:4])  # ★원하는 위치의 인덱스 값 (+ 1 한 인덱스) 0,1,2,3 ... 3 아니라 4
print("안녕하세요"[1:]) # 시작 인덱스
print("안녕하세요"[:3]) # 끝 인덱스
```

```python
# 문자 선택 연산자 [인덱스]
print("안녕하세요"[0:3])
print("안녕하세요"[:3]) # 콜론을 기준으로 처음 부터 카운팅 0,1,2
print("안녕하세요"[2:]) # 값이 없으면 마지막 까지 
```
