## 변수

```python
# 변수 : 데이터를 저장하기 위해 할당된 메모리 저장소 공간
print("3 + 5 = ", 3 + 5)
data1 = 3  # 데이터 저장 방법(equal 연산자 사용)
data2 = 5
print(data1 + data2 + 5) # 데이터 읽어오는 방법

 
```

```python
data1 = 7
data2 = 12
print(data1)
print("data1:" , data1)
print(data1 * data2)
```

```python
# 변수 선언과 할당과 저장
pi = 3.14159265
r = 10

# 변수 참조
print("원주율 =" , pi)
print("반지름 = " , r)
print("원의 둘레 = ", 2*pi*r)
print("원의 넓이 = ", pi*r*r)
```

## 복합 대입 연산자

```python
number = 100
number += 10
number += 20
number += 30
number -= 10  # 뺄셈 복합 대입 연산자
number *= 10
print("number =", number) # 누적이라는 단어의 함축적 의미!!
```

## input() 함수

```python
# 입력을 받습니다.
age = input("나이를 입력하세요>")
age = int(age)
print("35년 후의 내 나이는? :" , age + 35) # 반환값을 문자열로 지정했기 때문에 에러 발생
print("자료형:", type(age))
#name = input("이름을 입력하세요>")
# 출력합니다.
#print("이름:", name)
#print("자료형:", type(name))
```

```python
# 입력을 받습니다.
height = input("당신의 키는?>")
height = float(height)  # 실수형으로 변환
print("희망 키에 얼마가 모자라는 군요? :", int(198.80 - height))
print("희망 키에 얼마가 모자라는 군요? :", 198.80 - height)
print("자료형:", type(float))
```
