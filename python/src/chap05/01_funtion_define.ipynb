```python
# 기본적인 함수 정의

def gugudan():  # 함수를 구별하는 식별자(카멜, 스네이크 방식)
    for i in range(2, 10):   # 1 ~ 9
        for j in range(1, 10):
            print(f"{i} * {j} =  {i * j}")    
        print()
gugudan()  # 함수 호출 방법
```

```python
# 매개변수의 기본

def print_n_times(value, n):  # 두개값(Parameter 또는 매개변수)을 전달받음
    for i in range(n): # 0 ~ 4 range
        print(f"[{i}] {value}")

print_n_times("안녕하세요", 5)
```

```python
# 가변 매개변수 함수(변수명 앞에 *를 붙임)

def print_n_times(n, *values): # n 일반 매개변수, values 가변 매개변수
    # n번 반복합니다.
    for i in range(n): # 0부터 n - 1 까지 반복
        # values는 리스트처럼 활용
        for value in values:
            print(f"[values {i}] = {value}")
        # 단순한 줄바꿈
        print()

# 함수를 호출
print_n_times(3, "안녕하세요", "즐거운", "파이썬 프로그래밍")        
```

```python
# 가변 매개변수 뒤에는 일반 매개변수가 올 수 없다

def print_n_times_2(*values, n):
    # n번 반복한다
    for i in range(n):
        # values는 리스트처럼 활용
        for value in values:
            print(value)

        # 단순 줄바꿈
        print()    

print_n_times_2("아시아", "대한민국", "서울", 2) # 첫번째 매개변수 자리에 가변 매개변수 정의 -> TypeError발생
```

```python
# 범위 내부의 정수를 모두 더하는 함수
# 함수 선언
def sum_all(start, end):
    # 변수를 선언
    output = 0
    # 반복문을 돌려 숫자를 더함
    for i in range(start, end + 1):
        output += i
    # 리턴합니다
    return output

# 함수를 호출
print(f"0 to 100: {sum_all(0, 100)}")
print(f"0 to 1000: {sum_all(0, 1000)}")
print(f"50 to 100: {sum_all(50, 100)}")
print(f"500 to 100: {sum_all(500, 1000)}")
```

```python
# 기본 매개변수와 키워드 매개변수를 활용해 범위의 정수를 더하는 함수
# 기본 매개변수로 함수 선언

def sum_all(start=0, end=100, step=1):
    output = 0
    for i in range(start, end + 1, step):
        output += i
    return output    

# 함수를 호출
print(f"0 to 100: {sum_all()}")  # 키워드 매개변수, 호출시 우선순위가 가장 높다
print(f"(1): {sum_all(end=1000)}")
print(f"(2) 짝수의 총합: {sum_all(end=100, step=2)}")
```
