```python
# 튜플(tuple) 자료형 : 리스트와 유사,
# 리스트와 차이점은 한번 결정된 요소를 변경 할 수 없다
# tuple(list, dictionary, tuple데이터 등)
tuple_test = (10, 20, 30)
tuple_test[0] = 33 # 튜플은 내부 요소 변경이 불가능, 0번째 요소에 1을 넣으려고 하니 오류가 발생

print(tuple_test)
print(tuple_test[0])
```

```python
# 여러 개의 값 리턴하기

# 함수를 선언합니다.
def test():
    return (10, 20) # 튜플의 자료형 데이터
    # return [10, 20] # 리스트의 자료형으로 리턴해도 저장되는 값은 튜플과 동일, 리스트와 차이점은 한번 결정된 요소를 변경 할 수 없다 

# 여러 개의 값을 리턴받습니다.
a, b = test()  # a = 10, b = 20

# 출력합니다.
print("a:", a)
print("b:", b)
```

```python
# 함수의 매개변수로 함수 전달하기

# 매개변수로 받은 함수를 10번 호출하는 함수
def call_10_times(func):
    for i in range(10):  # 10번 반복
        func() # 함수 호출
        # pass # 기능의 완성을 나중에 하겠다

# 간단한 출력하는 함수
def print_hello():
    print("안녕하세요")

# 조합하기
call_10_times(print_hello)  # 매개변수에 함수 전달 def call_10_times(func)
```

```python
# map() 함수와 filter() 함수

# 함수를 선언합니다.
def power(item):
    return item * item

def under_3(item):
    return item < 3  # 비교연산 : 결과값은 boolean 값

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다.
output_a = map(power, list_input_a) # 두개의 매개변수 (함수와 값)을 전달받게 정의

print("# map() 함수의 실행 결과")
print("map(power, list_input_a):", output_a) # map 자료형을 가진 객체 반환 map object at
print("map(power, list_input_a):",list(output_a)) # map object -> list형으로 변환
print()

# filter() 함수를 사용합니다.
output_b = filter(under_3, list_input_a)
print("# filter() 함수의 실행 결과")
print("filter(under_3, list_input_a):", output_b)
print("filter(under_3, list_input_a):", list(output_b))
```

```python
# 람다 : 간단한 함수로 정의되어 있으면 lambda식으로 간단하게 정의

# 함수를 선언합니다.
# def power(item):
#     return item * item
power = lambda x: x * x  # def power(item):  return item * item 을 lambda 함수로 대체 가능

# def under_3(item):
#     return item < 3  # 비교연산 : 결과값은 boolean 값
under_3 = lambda x: x < 3  # x는 매개변수  def under_3(item): return item < 3 을 lambda 함수로 대체 가능

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다.
output_a = map(power, list_input_a) # 두개의 매개변수 (함수와 값)을 전달받게 정의

print("# lambda map() 함수의 실행 결과")
print("lambda map(power, list_input_a):", output_a) # map 자료형을 가진 객체 반환 map object at
print("lambda map(power, list_input_a):",list(output_a)) # map object -> list형으로 변환
print()

# filter() 함수를 사용합니다.
output_b = filter(under_3, list_input_a)
print("# lambda filter() 함수의 실행 결과")
print("lambda filter(under_3, list_input_a):", output_b)
print("lambda filter(under_3, list_input_a):", list(output_b))

 
```

```python
# 인라인 람다

# 변수를 선언합니다.
list_input_a = [1, 2, 3, 4, 5]

# map() 함수를 사용합니다.
output_a = map(lambda x: x * x, list_input_a) # 두개의 매개변수 (함수와 값)을 전달받게 정의

print("# 인라인 lambda map() 함수의 실행 결과")
print("인라인 lambda map(power, list_input_a):", output_a) # map 자료형을 가진 객체 반환 map object at
print("인라인 lambda map(power, list_input_a):",list(output_a)) # map object -> list형으로 변환
print()

# filter() 함수를 사용합니다.
output_b = filter(lambda x: x < 3, list_input_a)
print("# 인라인 lambda filter() 함수의 실행 결과")
print("인라인 lambda filter(under_3, list_input_a):", output_b)
print("인라인 lambda filter(under_3, list_input_a):", list(output_b))
 
```

```python
# 랜덤하게 1000명의 키와 몸무게 만들기

# 랜덤한 숫자를 만들기 위해 가져옵니다.
import random

# 간단한 한글 리스트를 만듭니다.
hanguls = list("가나다라마바사아자차카타파하") # 리스트 자료형으로 변환
    
# 파일을 쓰기 모드로 엽니다.
with open("info.txt", "w") as file:  # 옵션 "a" : 기존 데이터는 유지하면서 write할 데이터를 추가
    for i in range(100):
        # 랜덤한 값으로 변수를 생성
        name = random.choice(hanguls) + random.choice(hanguls)
        weight = random.randrange(40, 100)
        height = random.randrange(140, 200)
        # 텍스트를 파일에 저장
        file.write(f"{name}, {weight}, {height}\n") # , 단위
```

```python
# 반복문으로 파일 한 줄씩 읽기
with open("info.txt", "r") as file:
    for line in file:
        (name, weight, height) = line.strip().split(", ") # 콤마나 여백(스페이스, 탭, 줄바꿈이 ,로 분리) 튜플의 형태로 저장(데이터의 형태는 그대로 유지)

        # 데이터가 문제 없는지 검증하고 문제 있으면 지나감
        if (not name) or (not weight) or (not height):
            continue 

        # 결과를 계산
        bmi = int(weight) / ((int(height) / 100) ** 2)
        result = ""
        if 25 <= bmi:
            result = "과체중"
        elif 18.5 <= bmi:
            result = "정상 체중"    
        else:
            result = "저체중"    

        # 출력합니다
        print("\n".join([
            "이름 : {}",
            "몸무게 : {}",
            "키: {}",
            "BMI: {}",
            "결과 : {}"
        ]).format(name, weight, height, bmi, result))

        print()
```

```python
# 제너레이터 yield 함수

# 함수를 선언합니다.
def test():
    print("함수가 호출되었습니다.")
    yield test  # 제너레이터 함수, 이터레이터 객체 생성

# 함수를 호출합니다.
print("A 지점 통과")
test() # 제너레이트 함수 호출

print("B 지점 통과")
test()
print(test()) # 복귀할때 아무런 값이 없음 None 출력
```

```python
# 제너레이터 객체와 next() 함수

# 함수를 선언합니다. yield로 정의되면 test는 제너레이터 함수
def test():
    print("A 지점 통과") # 호출될 때 마다 yield를 만날때 까지 step별로 필요한 기능을 구현
    yield 1  # next를 만나면 return 처럼 동작, yield는 값만 반환하고 대기하고 있다가 다음 next를 만나면 해당 객체를 입력으로 전달 받았을 때 수행
    print("B 지점 통과")
    yield 2  # next함수를 호출해도 다음 yield 가 없는 경우 StopIteration 에러 발생                             
    print("C 지점 통과")
	
# 함수를 호출합니다.
output = test() # yield로 구성이 되어 있으면 제너레이터 객체를 반환(실행하는게 아니라 )
	
# next() 함수를 호출합니다.
print("D 지점 통과")
a = next(output) # 제너레이터 객체를 next 함수를 호출하면 비로소 함수 안으로 들어가게 됨
print(f"a {a}")	

print("E 지점 통과")
b = next(output)
print(f"b {b}")	

# print("F 지점 통과")
# c = next(output)
# print(f"c {c}")	

# 한번 더 실행하기
next(output)  # 더이상 실행할 이터레이터가 없으므로(yield의 갯수가 맞지 않음) StopIteration 에러 발생
```

```python
# 키워드 매개변수에 함수 전달하기
# 콜백 함수 : 파이썬에 의해 호출되어지는 함수
books = [{
    "제목": "혼자 공부하는 파이썬",
    "가격": 18000
}, {
    "제목": "혼자 공부하는 머신러닝 + 딥러닝",
    "가격": 26000
}, {
    "제목": "혼자 공부하는 자바스크립트",
    "가격": 24000
}]

def select_price(book):
    return book["가격"] # 리턴형을 보면 list, dictionary등이 입력값

print("# 가장 저렴한 책")
print(min(books, key=select_price)) # key라는 매개변수에 함수 전달  books는 dictionary 데이터, 기준이 모호한 경우 
print()

print("# 가장 비싼 책")
print(max(books, key=select_price)) # key라는 매개변수에 함수 전달
```

```python
# 콜백 함수를 람다로 바꾸기(callback 함수 : 파이썬에 의해 호출되는 특정 함수(min, max ....) 또는 기능)

books = [{
    "제목": "혼자 공부하는 파이썬",
    "가격": 18000
}, {
    "제목": "혼자 공부하는 머신러닝 + 딥러닝",
    "가격": 26000
}, {
    "제목": "혼자 공부하는 자바스크립트",
    "가격": 24000
}]

def select_price(book):
    return book["가격"] # 리턴형을 보면 list, dictionary등이 입력값

print("# 가장 저렴한 책")
print(min(books, key=lambda book:book["가격"])) # select_price 함수를 간단한 lambda 표현식으로 함수의 기능을 정의 (lambda 매개변수 선언, 리턴값 선언)
print()

print("# 가장 비싼 책")
print(max(books, key=lambda book:book["가격"]))   # max() 콜백 함수, min, max 처럼 기준값을 설정
```

```python
# 딕셔너리 오름차순 정렬하기

books = [{
    "제목": "혼자 공부하는 파이썬",
    "가격": 18000
}, {
    "제목": "혼자 공부하는 머신러닝 + 딥러닝",
    "가격": 26000
}, {
    "제목": "혼자 공부하는 자바스크립트",
    "가격": 24000
}]

print("# 가격을 기준으로 오름차순 정렬")
books.sort(key=lambda book:book["가격"]) # lambda 표현식으로 함수의 기능을 정의 (lambda 매개변수 선언, 리턴값 선언)
for book in books:
    print(f"{book}")

```
