# 1. 불 자료형과 if 조건문

```python
num1 = 20
num2 = 50

result = num1 > num2
print(result)

result = num1 < num2
print(result)

result = num1 <= num2
print(result)

result = (num1 == num2)
print(result)

result = (num1 != num2)
print(result)
```

### * 조건문의 기본 사용 (분기문)

```python
# 입력을 받습니다.
number = input("정수 입력> ") # input() 함수는 문자열을 리턴
number = int(number) # 형 변환

# 양수 조건
if number > 0:   # 콜론을 붙인다(조건의 끝이라는 의미)
    print("양수 입니다.")
    print("number :", number)

# 음수 조건
if number <  0:
    print("음수 입니다.")
    print("number :", number)

# 0 조건
if number == 0:
    print("0 입니다.")
    print("number :", number)

print("프로그램 종료")
# if 연산자는 비교 연산자와 함께 사용되어 true, false 값을 리턴
```

```python
# not 연산자 조합하기
x = 10
result = x < 20 
print('result :', result)
print("not result :", not result) # 단항 연산자
```

```python
# and 연산자 or 연산자 (논리 연산자)
X1 = False
X2 = True

output = X1 and X2
print("and :", output)

output = X1 or X2
print("or :", output)
```

### * 날짜/시간 활용하기

```python
# 날짜/시간과 관련된 기능을 가져옵니다.
import datetime

# 현재 날짜/시간을 구합니다.
now = datetime.datetime.now()

# 출력합니다.
print("{}년 {}월 {}일 {}시:{}분:{}초".format(
    now.year, # year는 함수가 아니고 변수
    now.month,
    now.day,
    now.hour,
    now.minute,
    now.second,
))

print(f"{now.year}년 {now.month}월 {now.day}일 {now.hour}시 {now.minute}분 {now.second}초") # f"{}" 문자열
#print(now.year, "년")
```

```python
# 날짜/시간과 관련된 기능을 가져옵니다.
import datetime

# 현재 날짜/시간을 구합니다.
now = datetime.datetime.now()

# 오전 구분
if now.hour < 12:
    print(f"현재 시각은 {now.hour}시 {now.minute}분이고 오전입니다")

# 오후 구분
if now.hour > 12:
    print(f"현재 시각은 {now.hour}시로 오후입니다")
```

```python
# 날짜/시간과 관련된 기능을 가져옵니다.
import datetime

# 현재 날짜/시간을 구합니다.
now = datetime.datetime.now()
month = now.month

# 봄 구분
if 3 <= month <= 5:
    season = "봄"
elif 6 <= month <= 8:
    season = "여름"
elif 9 <= month <= 11:
    season = "가을"    
else:
    season = "겨울"

print(f"이번달은 {month}월로 {season}입니다")
```

```python
import math

# 1. 사용자로부터 밑변과 높이 입력받기
a = float(input("밑변(a)의 길이를 입력하세요: "))
b = float(input("높이(b)의 길이를 입력하세요: "))

# 2. 피타고라스 정리를 이용해 빗변(c) 계산하기
c = math.sqrt(a**2 + b**2)

# 3. 결과 출력하기
print(f"빗변(c)의 길이는 {c} 입니다.")
```

```python
# 입력을 받습니다
number = input("정수 입력> ") # input()함수는 문자열로 반환

# 마지막 자리 숫자를 추출
last_character = number[-1] 

# 숫자로 변환하기
last_number = int(last_character)

# 1. 비교연산자 사용
if last_number == 0 \
    or last_number == 2 \
    or last_number == 4 \
    or last_number == 6 \
    or last_number == 8:
    print(f"{number}는 짝수입니다")  # 4칸 들여쓰기 필수

# 2. in 연산자 사용    
if last_character in "02468":  # find 함수와 동일하게 적용되는 연산자
    print(f"{number}는 짝수입니다")

if last_character in "13579":
    print(f"{number}는 홀수입니다")    

# 3. 나머지 연산자 %
# 몫을 반환하는 연산자 //
# 짝수 확인
if last_number % 2 == 0:
    print(f"{number}는 짝수입니다")
else:
    print(f"{number}는 홀수입니다")

divisor  = 2
quotient = int(number) // int(divisor)
print(f"{number}를 {divisor}(으)로 나눈 몫은 {quotient}입니다")
```
