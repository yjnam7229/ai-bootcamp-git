# 문제1) 문자열을 입력받아 모음(a e i o u)의 개수를 출력하세요.

```python
# 변수를 선언합니다.

example_list = ["a", "e", "i", "o", "u"]
output = []

# 입력받은 문자열(input_string)의 각 글자(char) 중에서 모음인 것만 골라 담습니다.
input_string = input("문자열 입력>")
output = [char for char in example_list if char in input_string]
# 리스트의 요소 개수가 곧 발견된 모음의 개수입니다.
cnt = len(output)
print(f"모음 리스트 {output} 개수 {cnt}")

```

## 문제2) 다음 딕셔너리가 있다. 다음을 출력하시오.
## - 평균 점수
## - 최고 점수 학생
## - 80점 이상 학생 이름



```python
scores = {
    "Kim": 80,
    "Lee": 92,
    "Park": 68,
    "Choi": 100
}

avg = 0
total_score = 0
max_score = -1
student = ""
score80 = []
# for 반복문과 items() 함수 조합해서 사용하기
for key, element in scores.items():
    total_score += element
    if element > max_score:
        max_score = element
        student = key
    if element >= 80:
       score80.append(key) 
print(f"합계 : {total_score}")
print(f"평균 : {total_score / len(scores)} 입니다")   
print(f"최고점수 학생은 {student} 입니다")
print(f"80점이상 학생 {score80}")
```

## 문제3) 다음 리스트에서

## numbers = [1,2,2,3,5,3,7,8,7,10]
## 중복을 제거한 뒤 오름차순으로 출력하시오.

```python
numbers = [1,2,2,3,5,3,7,8,7,10]

numbers.sort()
print(f"numbers {numbers}")

result = []

for num in numbers:
    if num not in result:
        result.append(num)
 
# 출력
print(f"제거된 후 value {result}")
```

## 문제4) 문장을 입력받아 
## I like python because python is fun
## 다음처럼 출력하시오.(딕셔너리 활용)
### I : 1
### like : 1
### python : 2
### because : 1
### is : 1
### fun : 1


```python
# I like python because python is fun 입력
input_string = input("문자열 입력> ")

words = input_string.split()
dictonary = {}  # 빈 딕셔너리 생성

for word in words:
    # print(f"word {word}")
    if word in dictonary:
        dictonary[word] += 1  # 💡 이미 있는 단어면 갯수 + 1
    else:
        dictonary[word] = 1   # 💡 처음 나온 단어면 1로 등록

# 결과 출력
for key, value in dictonary.items():
    print(f"{key} : {value}회")
```
