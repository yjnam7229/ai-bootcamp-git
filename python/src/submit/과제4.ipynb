### 문제2. 로또 번호 생성기를 프로그램으로 작성해 보세요.

### numbers = [12, 7, 25, 9, 18, 30]  합계 / 평균 / 최댓값 / 최솟값 / 짝수 개수 / 홀수 개수 구하기

```python
numbers = [12, 7, 25, 9, 18, 30]
# 합계
sum = 0
avg = 0
for num in  numbers:
    sum += num
print(f"합계 : {sum} 입니다")   

print(f"평균 : {sum / num} 입니다")   
output_avg = "{:g}".format(sum / num) 


# 최대값
numbers.sort()
max_num = numbers[-1]
print(f"최대값 {max_num}")

# 최소값
numbers.sort()
min_num = numbers[0]
print(f"최소값 {min_num}")

# 짝수 갯수
cnt1 = 0
for i in range(len(numbers)):
    # print(f"[{i}] {numbers[i]}")    
    if numbers[i] % 2 == 0:
        cnt1 += 1
print(f"짝수 갯수 {cnt1}개")    

# 홀수 갯수
cnt2 = 0
for j in range(len(numbers)):
    if numbers[j] % 2 != 0:
        cnt2 += 1
print(f"홀수 갯수 {cnt2}개")        
```
