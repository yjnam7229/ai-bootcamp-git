# 사용자에게 태어난 연도의 값을 입력받아, 띠를 출력하는 프로그램을 작성하세요.

```python
# 변수를 선언합니다.
# 12지신
list_a = ["원숭이", "닭", "개", "돼지", "쥐", "소", "호랑이", "토끼", "용", "뱀", "말", "양"]
year = int(input("태어난 연도>"))

zodiac = list_a[year % 12]

print(f"{year}년생은 {zodiac}띠입니다.")
```

## 연도를 입력받아 윤년인지 판별하세요.
### - 400으로 나누어 떨어지면 윤년  또는 4로 나누어 떨어지고 100으로는 나누어 떨어지지 않으면 윤년

```python
year = int(input("연도>"))

# if year % 400 == 0:
#     print(f"{year}년은 윤년입니다")
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year}년은 윤년 입니다")
else:
    print(f"{year}년은 윤년이 아닙니다")   
```

## 정수 3개를 입력받아 가장 큰 수를 출력하세요.

```python
a = int(input("정수 1>"))
b = int(input("정수 2>"))
c = int(input("정수 3>"))

max_number = a

if b > max_number:
    max_number = b

if c > max_number:
    max_number = c


print(f"가장 큰 수는 {max_number}입니다")
```
