## 문제1) 
## 좌표들이 저장된 튜플 리스트에서 x좌표의 합과 y좌표의 합을 구하세요.
## points = [(1,2), (3,4), (5,6)]
## 출력: x의 합 : 9
##       y의 합 : 12

```python
points = [(1,2), (3,4), (5,6)]

sumx = 0
sumy = 0

for x, y in points:
    # print(f"x 좌표 {x}, y 좌표 {y}")
    sumx += x
    sumy += y

print(f"x의 합 : {sumx}")
print(f"y의 합 : {sumy}")

```

# 문제3)
# 비밀번호는 다음 조건을 만족해야 합니다.
### - 길이 8자 이상
### - 숫자가 하나 이상 포함
### - 영문자가 하나 이상 포함
### 조건을 만족하면 "사용 가능한 비밀번호입니다."
### 그렇지 않으면 "사용할 수 없는 비밀번호입니다." 를 출력하세요.



```python
password = input("문자열 입력>")

# 길이 조건
length = len(password) >= 8

has_digit = len(list(filter(lambda char: char.isdigit(), password))) > 0
has_alpa = len(list(filter(lambda char: char.isalpha(), password))) > 0

if length and has_digit and has_alpa:
    print("사용 가능한 비밀번호입니다.")
else:
    print("사용 할 수 없는 비밀번호입니다.")    
```

## 문제4)
### 숫자를 입력받아 소수인지 판별하는 함수를 작성하세요.
### 입력 17
### 출력 소수입니다.


```python
input_num = int(input("문자열 입력>"))

def is_prime(num):
    if num <= 1:
        return False

    # 2부터 (num - 1)까지의 숫자로 나누어 떨어지는지 검사
    for i in range(2, num - 1):
        if num % i == 0:
            return False
    return True

if is_prime(input_num):
    print(f"{input_num} 는 소수입니다")
else:
    print(f"{input_num} 는 소수가 아닙니다")
```

## 문제5)
### words = ["python", "java", "c", "javascript"]
### 각 단어의 길이를 하나씩 반환하는 제너레이터를 작성하세요.
### 출력  
      6 
      4 
      1 
      10

```python
words = ["python", "java", "c", "javascript"]

# 단어의 길이를 생성하는 제너레이터
length_text = (len(val) for val in words) # 소괄호 제너레이터
print(f"length_text {list(length_text)}\n") # 리스트로 변환한 결과를 f-string 내부에서 출력하고 뒤에 줄바꿈 추가

# yield 함수 사용
def words_get(length_words):
    for val in length_words:
        yield len(val)
        
length_text = words_get(words)

# words_get(words)로 생성된 제너레이터 알맹이들을 *로 탈탈 털어내고,
# 줄바꿈(\n)을 사이에 끼워 세로로 출력합니다.
print(*length_text, sep="\n")
 
```

```python
words = ["python", "java", "c", "javascript"]

# 1. 제너레이터 함수 정의
def get_lengths(word_list):
    for val in word_list:
        yield len(val)

# 2. 제너레이터 객체 생성
length_gen = get_lengths(words)

# 3. 언패킹 후 줄바꿈(\n)을 기준으로 출력
print(*length_gen, sep="\n")
```
