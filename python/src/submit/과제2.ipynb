## 피타고라스의 정리
     a^2 + b^2 = c^2 에서
사용자로 부터 밑변의 길이와, 높이의 값을 입력 받았을 때, 
빗변의 길이를 출력하는 프로그램

```python
a = float(input("밑변(a)의 길이를 입력하세요: "))
b = float(input("높이(b)의 길이를 입력하세요: "))

result = (a**2 + b**2) ** 0.5
print(f"빗변의 길이는 {result} 입니다")
```

```python
import math

# 1. 사용자로부터 밑변과 높이 입력받기
a = float(input("밑변(a)의 길이를 입력하세요: "))
b = float(input("높이(b)의 길이를 입력하세요: "))

# 2. 피타고라스 정리를 이용해 빗변(c) 계산하기
c = math.sqrt(a**2 + b**2)

# 3. 결과 출력하기
print(f"빗변(c)의 길이는 {c} 입니다.")
```
