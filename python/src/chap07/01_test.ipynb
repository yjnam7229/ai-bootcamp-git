## random 모듈 : 값을 랜덤하게 생성

```python
import random as rnd

print("# random 모듈")

# 0.0 <= x < 1.0 사이의 float를 리턴
print("- random():", rnd.random())

# uniform(min, max) : 지정한 범위 사이의 float를 리턴
print("- uniform(10, 20)", rnd.uniform(10, 20))

# randrange(): 지정한 범위의 int를 리턴 [기본 매개변수로 정의되어 있는 함수는 변수에 값을 전달하지 않아도 됨, 한개, 두개, 빈 값이어도 됨]
# randrange(max): 0 부터 max 사이의 값을 리턴
# randrange(min, max) : min 부터 max 사이의 값을 리턴
print("- randrange(10, 100)", rnd.randrange(10,100)) 
print("- randrange(10)", rnd.randrange(10)) 


# shuffle(list): 리스트의 요소를 랜덤하게 섞습니다.
shuffle_data = [1, 2, 3, 4, 5]
print("- shuffle[1, 2, 3, 4, 5]:", rnd.shuffle(shuffle_data))
print(shuffle_data)

# sample(list, k=<숫자>): 리스트의 요소 중에 k개를 뽑습니다.
print("- sample([1, 2, 3, 4, 5], k=2):", rnd.sample([1, 2, 3, 4, 5], k=2))

# 1부터 45까지의 숫자 리스트 만들기
number_list = list(range(1, 46))
row_num = rnd.sample(number_list, 6)
print("로또번호 :", row_num)

# number_range = rnd.randrange(1, 46)
# print("로또번호 :", number_range)
```

## datetime 모듈

```python
import datetime

print("# 현재 시각 출력하기")
now = datetime.datetime.now() 

print(now.year, "년")
print(now.month, "월")
print(now.day, "일")
print(now.hour, "시")
print(now.minute, "분")
print(now.second, "초")
print()

# 시간 출력
print("# 시간을 포맷에 맞춰 출력하기")
output_a = now.strftime("%Y.%m.%d %H:%M:%S")
output_b = "{}년 {}월 {}일 {}시 {}분 {}초".format(now.year,\
        now.month,\
        now.day,\
        now.hour,\
        now.minute,\
        now.second)

output_c = now.strftime("%Y{} %m{} %d{} %H{} %M{} %S{}").format(*"년월일시분초")   # * 전계 연산자 사용, %m을 사용하면 빈자리는 0을 채워서 출력

print(output_a)
print(output_b)
print(output_c)

print()
```

```python
import datetime

now = datetime.datetime.now()
# 특정 시간 이후의 시간 구하기
print("# datetime.timedelta로 시간 더하기")
# 현재를 기준으로 weeks + 1, day + 1, hours + 1 ,,,, 
after = now + datetime.timedelta(\
    weeks=1,\
    days=1,\
    hours=1,\
    minutes=1,\
    seconds=1)
print(after.strftime("%Y{} %m{} %d{} %H{} %M{} %S{}").format(*"년월일시분초"))

print()

# 특정 시간 요소 교체하긱
print("# now.replace()로 1년 더하기")
output = now.replace(year=(now.year + 1))
print(output.strftime("%Y{} %m{} %d{} %H{} %M{} %S{}").format(*"년월일시분초"))
```

```python
class Students:
    # 생성자 함수: 객체 생성 시 파이썬이 자동으로 호출하는 콜백 함수
    def __init__(self, name, korean, math, english, science):
        self.name = name
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science

    # 총점 계산 메서드
    def get_sum(self):
        return self.korean + self.math + self.english + self.science
    
    # 평균 계산 메서드
    def get_average(self):
        return self.get_sum() / 4
    
    # 문자열 반환 메서드: print()나 str() 함수를 만났을 때 자동으로 호출됨
    def __str__(self):
        return f"{self.name}\t{self.get_sum()}\t{self.get_average():.2f}"


# 1. 여러 학생 객체(인스턴스)들을 생성하여 하나의 리스트에 담습니다.
student_list = [
    Students("홍길동", 90, 79, 95, 80),
    Students("이순신", 85, 90, 88, 92),
    Students("유관순", 95, 100, 92, 96)
]

# 2. 상단 타이틀을 먼저 출력합니다.
print("이름", "총점", "평균", sep="\t")
print("-" * 30)

# 3. 학생 리스트에서 한 명씩 꺼내어 정보를 출력합니다.
for student in student_list:
    # str(student)를 실행하면 클래스 내부의 __str__ 함수가 알아서 작동합니다.
    print(str(student))
```

```python
class Students:
    def __init__(self, name, korean, math, engish, science):  # 매개변수 반드시 선언, 파이썬에 의해 호출되는 콜백 함수
        self.name = name
        self.korean = korean
        self.math = math
        self.engish = engish
        self.science = science

    # 총점 계산
    def get_sum(self):
        return self.korean + self.math + self.engish + self.science
    
    # 평균 계산
    def get_average(self):
        return self.get_sum() / 4
    
    def __str__(self):
        return (f"{self.name}\t {self.get_sum()}\t {self.get_average():.2f}")
    

# 인스턴스 확인
# print(f"isIncstance(student, Student) {isinstance(student, Student)}")

# students = Students()  # 반드시 클래스의 이름와 동일하게 호출(생성자 함수) __init__()함수 호출

# 학생 리스트를 선언합니다.
students = [
        Students("홍길동", 90, 79, 95, 80)
]
print("이름", "총점", "평균", sep="\t")

# 학생을 한 명씩 반복합니다.
for student in students:
    print(str(student))
```

```python
class Students:
    # 생성자 함수: 객체 생성 시 데이터를 변수에 안전하게 저장
    def __init__(self, name, korean, math, english, science):
        self.name = name
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science

    # 총점 계산 메서드
    def get_sum(self):
        return self.korean + self.math + self.english + self.science
    
    # 평균 계산 메서드
    def get_average(self):
        return self.get_sum() / 4
    
    # 표 칸에 딱 맞게 정렬하여 문자열을 반환하는 메서드
    def __str__(self):
        # <6 은 6칸을 확보하고 왼쪽 정렬, >8 은 8칸을 확보하고 오른쪽 정렬을 뜻합니다.
        return f"{self.name:<4}\t{self.korean:>8}\t{self.math:>8}\t{self.english:>8}\t{self.science:>8}"


# 1. 학생 객체들을 리스트로 생성합니다.
student_list = [
    Students("홍길동", 90, 79, 95, 80),
    Students("이순신", 85, 92, 78, 88),
    Students("유관순", 95, 100, 92, 96)
]

# 2. 상단 타이틀 양식 출력
print("-" * 70)
print(f"{'이름':<4}\t{'국어점수':>8}\t{'수학점수':>8}\t{'영어점수':>8}\t{'과학점수':>8}")
print("-" * 70)

# 3. 반복문을 돌며 학생 데이터 출력
for student in student_list:
    print(str(student))
    print(f"{student.name}\t    {student.korean}\t    {student.math}\t    {student.engish}\t    {student.science}")

```
