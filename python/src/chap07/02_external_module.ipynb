## 함수 데코레이터 

```python
# 함수 데코레이터를 생성합니다.
def test(function):
    def wrapper():
        print("인사가 시작되었습니다.")
        function()
        print("인사가 종료되었습니다.")
    return wrapper

# 데코레이터를 붙여 함수를 만듭니다.
@test
def hello():
    print("hello")

# 함수를 호출합니다.
hello()
```

```python
# 함수 데코레이터를 생성

def test(function):  # @test 데코레이터 함수에 의해 hello 함수를 입력으로 넣어준다
    def wrapper():
        print("인사가 시작되었습니다.")
        function()   # hellow 함수 호출
        print("인사가 종료되었습니다.")   
    return wrapper

# 데코레이터를 붙여 함수를 만듭니다.
@test
def hello():
    print("hello")

# 함수를 호출합니다
hello()
```

## 데코레이터 함수 태스트

```python
# 데코레이터 정의
def logger_decorator(original_function):
    # *args, **kwargs를 사용하여 어떤 매개변수든 다 받아줍니다
    def wrapper(*args, **kwargs):
        print(f"[로그] {original_function.__name__} 함수가 실행됩니다.")
        
        # 원래 함수를 실행하고 결과값을 저장합니다
        result = original_function(*args, **kwargs)
        
        print(f"[로그] {original_function.__name__} 함수가 종료되었습니다.")
        return result  # 결과값을 돌려줍니다
    return wrapper

# 데코레이터 적용
@logger_decorator
def add_numbers(a, b):
    return a + b

# 함수 실행
total = add_numbers(5, 7)
print(f"결과값: {total}")
```
