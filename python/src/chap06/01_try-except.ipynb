```python
# 조건문으로 예외 처리하기

# 숫자를 입력받습니다.
user_input_a = input("정수 입력> ")

# 사용자 입력이 숫자로만 구성되어 있을 때
if user_input_a.isdigit():
    number_input_a = int(user_input_a)
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)

else:
    print("정수를 입력하지 않았습니다.")
```

```python
# try except 예외 처리하기

# 숫자를 입력받습니다.
user_input_a = input("정수 입력> ")

# 사용자 입력이 숫자로만 구성되어 있을 때
try:
    number_input_a = int(user_input_a)
    
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)

except: # 예외인 경우
     print("정수를 입력하지 않았습니다.")
    #  pass 에러를 내지 않으면서 기능구현을 잠시 보류
```

# try except 구문으로 예외를 처리합니다.



```python
# 숫자로 변환되는 것들만 리스트에 넣기

# 변수를 선언합니다.
list_input_a = ["52", "273", "32", "스파이", "103"]

# 반복을 적용합니다.
list_number = []
for item in list_input_a:
    try:
        float(item)
        list_number.append(item)
    except:
        pass

# 출력합니다.
print("{} 내부에 있는 숫자는".format(list_input_a))
print("{} 입니다".format(list_number))
```

```python
# try ~ except ~ else 구문으로 예외를 처리합니다.

# 숫자를 입력받습니다.
user_input_a = input("정수 입력> ")

# 사용자 입력이 숫자로만 구성되어 있을 때
try:
    number_input_a = int(user_input_a)
except: # 예외인 경우
     print("정수를 입력하지 않았습니다.")
else: # 예외가 발생되지 않았을 때
    # 출력합니다  
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)
finally:
    print("일단 프로그램이 어떻게든 끝났습니다.")    
```

```python
# finally 구문

# 숫자를 입력받습니다.
user_input_a = input("정수 입력> ")

# 사용자 입력이 숫자로만 구성되어 있을 때
try:  # 콜론의 의미는 명령문이 다음으로 연결되어 있다
    number_input_a = int(user_input_a)
except: # 예외인 경우
     print("정수를 입력하지 않았습니다.")
finally: # 예외가 발생되지 않았을 때
    # 출력합니다  
    print("원의 반지름:", number_input_a)
    print("원의 둘레:", 2 * 3.14 * number_input_a)
    print("원의 넓이:", 3.14 * number_input_a * number_input_a)
```

```python
def test():
    print("test() 함수의 첫 줄입니다.")

    try:
        print("try 구문 실행")
        return # 함수를 벗어나서 그 함수를 호출한 쪽으로 복귀, 복귀하면서 수행하고난 결과를 return뒤에 담아서 복귀하는 의미 
        print("try 구문의 return 키워드 뒤입니다.")
    except:
        print("except 구문 실행")
    else:
        print("else")
    finally: # 함수안의 return을 만난다고 하더라도 무조건 호출해서 복귀한다
        print("finally 구문 실행")

    print("test() 함수의 마지막 줄입니다.")

# test() 함수를 호출합니다.
test() # return을 만나면 호출한 쪽으로 복귀
 
```

```python
# 반복문과 함께 사용하는 경우

print("프로그램이 시작되었습니다.")

while True: 
    try:
        print("try 구문이 실행되었습니다.")
        break 
        print("try 구문의 return 키워드 뒤입니다.")
    except:
        print("except 구문 실행")
    else:
        print("else")
    finally: # 함수안의 break 만난다고 하더라도 무조건 호출해서 복귀한다
        print("finally 구문 실행")

print("test() 함수의 마지막 줄입니다.")

print("프로그램이 종료되었습니다.")
```
