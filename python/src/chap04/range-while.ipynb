## 3. 범위 자료형과 while 반복문

#### * 범위

```python
print(range(5))
print(range(1, 5))
print(range(1, 10, 2))
```

```python
# range : 정수 값을 생성하는 함수
step = range(5)  # range(0, 5)  인덱스 0부터 시작해서 직전 값까지 생성 (0, 1, 2, 3, 4), 슬라이싱 처럼 마지막 인덱스 -1 까지
list_value = list(step) # 입력으로 전달받은 step 데이터를 리스트 형으로 변환 : list()함수
print(list_value)


step = range(1, 5)   
list_value1 = list(step) # 입력으로 전달받은 step 데이터를 리스트 형으로 변환 : list()함수
print(list_value1)

step = range(1, 10, 2)   # 3번째 값은 간격 2개씩 건너뜀
list_value2 = list(step) # 입력으로 전달받은 step 데이터를 리스트 형으로 변환 : list()함수
print(list_value2)

# list_a = list(range(1, 10, 2))
# print(list_a)

 
```

```python
# for 반복문과 범위

# for 반복문과 범위를 함께 조합해서 사용합니다.
for i in range(5): # 반복의 의미만 담아서 수행 range()함수 사용
    print(f"{i} = 반복 변수")

print()
for i in range(5, 10):
    print(f"{i} = 반복 변수")

print()
for i in range(0, 10, 3):
    print(f"{i} = 반복 변수")
```

```python
# 리스트와 범위를 조합해서 사용하기

# 리스트를 선언합니다.
array = [273, 32, 103, 57, 52]

# 리스트에 반복문을 적용합니다.
for i in range(len(array)):
    print(f" [{i}] 번째 값 = {array[i]}")
```

```python
# 역 반복문
for i in range(4, 0 - 1, -1):  # 0 - 1 직전 값이 0까지 카운팅됨, 최종적으로 수행이 완료된 하나의 정수값
    print(f"현재 반복 변수 : {i}")

print()
# 반대로 반복하기
for i in reversed(range(0, 5)):
    print(f"현재 반복 변수 : {i}")    
```

```python
# 반복문으로 피라미드 만들기

output = ""
for i in range(1, 10):  # 9번 반복
    for j in range(0, i):  # 0, 1  0, 2  0, 3 ...
         output += "*"
    output += "\n"

print(output)
```

```python
output = ""

for i in range(1, 10):
    for j in range(9, i, -1):
        output += ' '
    for k in range(0, 2 * i - 1):
        output += "*"
    output += "\n" 

print(output)      
```
