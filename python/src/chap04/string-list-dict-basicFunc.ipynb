```python
# 무한반복

# while 반복문을 사용합니다
while True:
    # .을 출력합니다.
    # 기본적으로 end가 "\n"이라 줄바꿈이 일어나는데
    # 빈 문자열 ""로 바꿔서 줄바꿈이 일어나지 않게 합니다
    print(".", end="")
```

```python
# while 반복문을 for 반복문처럼 사용하기
# 반복 변수를 기반으로 반복하기
i = 0
while i < 10:
    # print(f"[{i}]번째 반복입니다.")
    print(f"[{i}]번째 반복입니다   ", end="") # 빈 문자열 ""로 바꿔서 줄바꿈이 일어나지 않게 합니다
    i += 1
    
print()
print(f"while문 빠져나오는 [{i}] 입니다")   
```

```python
# 해당하는 값 모두 제거하기

# 변수를 선언
list_test = [1, 2, 1, 2]
value = 2

# list_test 내부에 value가 있다면 반복
while value in list_test:  # list 또는 dictionary 올 수 있음, while문의 in은 연산자
    list_test.remove(value)

# 출력
print(f"제거된 후 value {list_test}")
 
```

```python
# 성능 체크 : time.time() 함수를 이용
# 5초 동안 반복하기
# 시간과 관련된 기능을 가져옴
import time

# UTC(셰계 표준시)로 1970년 1월 1일0시 0분 0초를 기준으로 실행되는 순간을 가져와서 몇 초가 지났는지 정수로 나타낸 것
print(time.time())

# 변수 선언
number = 0

# 5초 동안 반복
target = time.time() + 5 # 실행한 순간의 시간 + 5초
while time.time() < target:
    number += 1

# 출력
print("5초 동안 {} 번 반복했습니다".format(number))
 
```

```python
# break 키워드
# 변수 선언
i = 0

# 무한 반복
while True:
    print(f"{i}번째 반복힙니다.")
    i += 1
    # 반복을 종료합니다
    input_text = input("> 종료하시겠습니까(y):") # 실시간으로 입력
    if input_text in ["y", "Y"]:
        print("반복을 종료합니다.")
        break

print(f"마지막 종료 {i}")
```

```python
# continue 키워드
# 변수 선언
numbers = [5, 15, 6, 20, 7, 25]
# 반복을 돌립니다
for num in numbers:
# for num in range(len(numbers)):    
    # num 값이 10보다 작으면 다음 반복으로 넘어갑니다.
    if num < 10:
        continue
    # 출력
    print(f"num 값 {num}")
```

```python
# reversed() 함수로 리스트 뒤집습니다

# 리스트를 선언하고 뒤집습니다.
list_a = [1, 2, 3, 4, 5]
list_reversed = reversed(list_a)

# 출력합니다.
print("# reversed() 함수")
print("reversed([1, 2, 3, 4, 5]):", list_reversed) 
print("list(reversed([1, 2, 3, 4, 5])):", list(list_reversed)) # list 자료형으로 변환
print()

# 반복문을 적용해 봅니다.
print("# reversed() 함수와 반복문")
print("for i in reversed([1, 2, 3, 4, 5]):")
for i in reversed(list_a):
    print("*", i)
```

```python
# enumerate() 함수와 리스트

# 변수를 선언합니다.
example_list = ["요소A", "요소B", "요소C"]

# 그냥 출력합니다.
print("# 단순 출력")
print(example_list)
print()

# enumerate() 함수를 적용해 출력합니다.
print("# enumberate() 함수 적용 출력")
print(enumerate(example_list))
print()

# list() 함수로 강제 변환해 출력합니다.
print("# list() 함수로 강제 변환 출력")
print(list(enumerate(example_list))) # list 담긴 값에 번호를 부여해서 접근
print()

# for 반복문과 enumerate() 함수 조합해서 사용하기
print("# 반복문과 조합하기")
for i, val in enumerate(example_list):  # enumerate은 두개의 값을 반환
    print(f"{
        
        i}번째 값은 {val} 입니다")

```

```python
# 딕셔너리의 items() 함수와 반복문 조합하기

# 변수를 선언합니다.
example_dictionary = {
    "키A": "값A",
    "키B": "값B",
    "키C": "값C",
}

# 딕셔너리의 items() 함수 결과 출력하기
print("# 딕셔너리의 items() 함수")
print("items():", example_dictionary.items())
print()

# for 반복문과 items() 함수 조합해서 사용하기
print("# 딕셔너리의 items() 함수와 반복문 조합하기")

for key, element in example_dictionary.items():
    print(f"dictionary [{key}] = {element}")
```

```python
# 반복문을 사용한 리스트 생성

# 변수를 선언합니다.
array = []

# 반복문을 적용합니다.
for i in range( 0, 20, 2): # 시작 0, 끝 19, step 2씩 증가
    array.append(i * i) # 마지막에 추가

# 출력합니다.
print(array)
```

```python
# 리스트 안에 for문 사용하기

# 리스트를 선언합니다.
arrary = [i * i for i in range(0, 20, 2)] # list comprehensions로 명칭, list안에서 내포되는 표현식, list comprehension 내포

# 출력합니다.
print(arrary)
```

```python
# 조건을 활용한 리스트 내포(list comprehensions)

# 리스트를 선언합니다.
array = ["사과", "자두", "초콜릿", "바나나", "체리"]
# output = [fruit for fruit in arrary if fruit != '']

output = []
for fruit in array:
    if fruit != "초콜릿":
        output.append(fruit)

print("첫번째 ", output)

# list comprehensions 리스트 데이터의 element들을 생성
output = [fruit for fruit in array if fruit != "초콜릿"]

# 출력합니다.
print("두번째 ", output)
```

```python
# if 조건문과 여러 줄 문자열

# 변수를 선언합니다.
number = int(input("정수 입력> ")) # 문자열로 반환되므로 정수값으로 변환

# if 조건문으로 홀수 짝수를 구분합니다.
if number % 2 == 0:
    print("""\
            입력한 문자열은 {}입니다.
            {}값은(는) 짝수입니다.
          """.format(number, number))
else:
    print("""\n
            입력한 문자열은 {}입니다.
            {}값은(는) 홀수입니다.
          """.format(number, number))
```

```python
# 변수를 선언합니다.
number = int(input("정수 입력> "))

# if 조건문으로 홀수 짝수를 구분합니다.
if number % 2 == 0:
    print("\n".join([
         "입력한 문자열은 {}입니다.",   # 콤마 중요
        "{}은(는) 짝수입니다"
    ]).format(number, number))
else:
    print("\n".join([
        "입력한 문자열은 {}입니다",
        "{}은(는) 홀수입니다"
    ]).format(number, number))
```
