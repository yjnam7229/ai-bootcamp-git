## 2. 딕셔너리와 반복문
### 데이터를 key, value 형태로 관리

```python
# 딕셔너리의 요소에 접근하기

# 딕셔너리를 선언합니다.
dictonary = {
    "name": "7D 건조 망고", 
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중화황산나트륨", "치자황색소"],
    "orgin": "필리핀"
}  # key, value

# 출력합니다.
print("name:", dictonary["name"])
print("type:", dictonary["type"])
print("ingredient:", dictonary["ingredient"])
print("orgin:", dictonary["orgin"])
# print(dictonary.keys())
print(f"{dictonary.keys()}\n{dictonary.values()}")

# 값을 변경합니다.
dictonary["name"] = "8D 건조 망고"
print(dictonary["name"])
```

```python
# 딕셔너리 요소 추가하기

# 딕셔너리를 선언합니다.
dictonary = {} # 초기화

# 요소 추가 전에 내용을 출력해 봅니다.
print("요소 추가 이전:", dictonary)

# 딕셔너리에 요소를 추가합니다.
dictonary["name"] = "홍길동"
dictonary["age"] = "20"
dictonary["addr"] = "서울시 종로구"

print("요소 추가 이후", dictonary)
print("나이", dictonary["age"])
```

```python
# 딕셔너리 선언
dictonary = {
    "name": "7D 건조 망고", 
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중화황산나트륨", "치자황색소"],
    "orgin": "필리핀"
}
# 요소 추가 전에 내용을 출력해 봅니다.
print("요소 추가 이전:", dictonary)

# 딕셔너리의 요소를 제거합니다.
del dictonary['ingredient']
print("요소 제거 이후:", dictonary)
```

```python
# 키가 존재하는지 확인하고 값에 접근하기
dictonary = {
    "name": "7D 건조 망고", 
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중화황산나트륨", "치자황색소"],
    "orgin": "필리핀"
}

# 사용자로부터 입력을 받습니다.
key = input("> 접근하고자 하는 키: ")

# 출력합니다.
if key in dictonary:  # in은 연산자로 동작 되어짐(존재 유무 True/False)
    print(dictonary[key])
else:
    print("존재하지 않은 키에 접근하고 있습니다.")
```

```python
# 키가 존재하지 않을 때 None을 출력하는지 확인하기

# 딕셔너리를 선언합니다.
dictonary = {
    "name": "7D 건조 망고", 
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중화황산나트륨", "치자황색소"],
    "orgin": "필리핀"
}

# 존재하지 않는 키에 접근해 봅니다.
value = dictonary.get("age")
# value = dictonary.get("name")

print(f"값 = {value}")

# None 확인 방법
if value == None: # 존재하지 않으면
    print("존재하지 않은 키에 접근")
```

```python
# for 반복문과 딕셔너리

# 딕셔너리를 선언합니다.
dictonary = {
    "name": "7D 건조 망고", 
    "type": "당절임",
    "ingredient": ["망고", "설탕", "메타중화황산나트륨", "치자황색소"],
    "orgin": "필리핀"
}

# for 반복문을 사용합니다.
for key in dictonary:  # key, value형식으로 데이터가 존재
    # print(key) # 키 값만 가져옴
    # print(key, ":" , dictonary[key])  # value에 해당하는 데이터 출력
    print(f"{key} {dictonary[key]}")
    
```
