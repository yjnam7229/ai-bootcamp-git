```python
import pandas as pd

metting_note_csv_path = '../audio/싼기타_비싼기타_final.csv'
df_rttm = pd.read_csv(metting_note_csv_path, sep='|')

display(df_rttm)
```

```python
# 이름 넣기
name_dict = {
    'SPEAKER_00': 'AI',
    'SPEAKER_01': '홍길동'
}

df_rttm['name'] = df_rttm['speaker_id'].apply(lambda x: name_dict[x])
display(df_rttm)
```

```python
meeting_note_txt = df_rttm[['start', 'end', 'name', 'text']].to_json(orient='records', force_ascii=False)   # orce_ascii=False 문자열 그대로 변환
print(meeting_note_txt)
```

```python
system_prompt = f'''
    너는 회의 내용을 요약하는 봇이야. 아래 회으록을 읽고, 주요 내용을 요약하라.
    결과는 마크다운 형식으로 작성한다.

    아래 형식에 맞추어 작성하라.

    # 회의 제목

    ## 주요 내용

    ## 참석자별 입장

    ## 결정 사항

    ================== 이하 회의록 ====================

    {meeting_note_txt}

'''
```

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)
```

```python
# OpenAI API를 이용해 요약 결과 생성
response= client.chat.completions.create(
    model= 'gpt-5-nano',
    messages=[
    {"role": "system", "content": system_prompt},
    ]
)

# 요약 결과를 파일로 저장
summary = response.choices[0].message.content
summary = summary.strip()  # 좌우 공백 제거

print(summary)
```

```python
with open('../audio/guitar_summary.md', 'w', encoding='utf-8') as f:
    f.write(summary)
```

```python
df_meeting_note = df_rttm[['start', 'end', 'name', 'text']].copy()
df_meeting_note.dropna(inplace=True) # 항목에 데이터가 담겨저 있지 않으면 제거

meeting_note_dict = df_meeting_note.to_dict(orient='records') # 딕셔너리 형태로 변환

meeting_note_dict   
```

```python
import re
from openai import OpenAI

# OpenAI 클라이언트 초기화 (필요시 api_key 전달)
client = OpenAI()

# 1. 공통 context 및 system prompt 사전 준비
system_prompt = """너는 주어진 회의 녹취록을 수정 및 보완하는 전문가이다.
            제시된 '회의 요약문'과 '전체 녹취록'의 문맥을 참조하여, '확인할 메시지 원본'의 STT 오탈자나 문맥상 오류를 수정하라.

            [수정 규칙]
            1. 표현과 말투는 최대한 원본과 일치하도록 유지하라.
            2. 잘못 받아적은 내용, 오탈자, 불명확한 단어만 문맥에 맞게 수정하라.
            3. 설명, 이유, 서론/본론 구분 문구, '수정된 내용:' 같은 접두사 없이 **오직 수정된 메시지 텍스트만** 반환하라.
            """

# 전체 녹취록 JSON 문자열은 루프 밖에서 1회만 변환하여 재사용
meeting_note_json = df_meeting_note.to_json(orient='records', force_ascii=False)

# 2. 메시지별 보정 루프
for row in meeting_note_dict:
    # 텍스트가 아니거나 빈 값인 경우 건너뛰기
    if not isinstance(row.get('text'), str) or not row['text'].strip():
        continue

    user_prompt = f"""--------------------------------
    [회의 요약문]
    {summary}

    --------------------------------
    [회의 녹취록 전문]
    {meeting_note_json}

    --------------------------------
    [확인할 메시지 원본]
    {row['text']}"""

    response = client.chat.completions.create(
        model='gpt-5-nano',  # 사용 중이신 모델명 지정 (예: gpt-4o-mini, gpt-4o 등)
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        # temperature=0.2  # 일관되고 정확한 수정을 위해 낮은 온도 설정
    )

    correction = response.choices[0].message.content.strip()

    # 안전한 접두사 제거 (콜론이 본문에 포함되어도 문제 없도록 정규식 활용)
    correction = re.sub(r'^(수정|보완|결과|응답)\s*:\s*', '', correction)
    correction = correction.strip()

    # 결과 확인 및 저장
    print(f"화자: {row.get('name', 'Unknown')}")
    print(f"원본: {row['text']}")
    print(f"수정: {correction}")
    print('-' * 40)

    row['corrected_text'] = correction
```

```python
# STT 원본을 GPT로 교정하기
# meeting_note_dict를 JSON 파일 형식으로 저장

import json

with open('../audio/guitar_meeting_note_corrected.json', 'w', encoding='utf-8') as f:
    json.dump(meeting_note_dict, f, ensure_ascii=False, indent=4)
```

```python
# guitar_meeting_note_corrected.json 파일을 마크다운 형식으로 저장

md_template = ''

for row in meeting_note_dict:
    md_template += f"- **{row['name']}** : {row['corrected_text']}\n"

with open('../audio/guitar_meeting_note_corrected.md', 'w', encoding='utf-8') as f:
    f.write(md_template)
```

```python
from docx import Document

document = Document()                       # (1) docx 객체 생성
document.add_heading('회의록', level=1)     # (2) 제목 추가
table = document.add_table(rows=1, cols=2)  # (3) 테이블

table.style = 'Table Grid'      # (4) 테이블 스타일 지정
hdr_cells = table.rows[0].cells # (5) 첫 번째 행에 셀 추가
hdr_cells[0].text = 'Speaker'   # (6) 첫 번째 셀에 Speaker 추가
hdr_cells[1].text = 'Content'   # (7) 두 번째 셀에 Content 추가

# (8) 수정된 녹취록 파일을 읽어들여서 테이블에 추가
for row in meeting_note_dict:
    row_cells = table.add_row().cells
    row_cells[0].text = row['name']
    row_cells[1].text = row['corrected_text']

document.save('../audio/guitar_meeting_note_corrected.docx')    # (9) docx 파일 저장
```
