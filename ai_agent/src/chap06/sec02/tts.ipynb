```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)  # OpenAI 객체의 인스턴스를 생성
```

```python
response = client.audio.speech.create(
    model='tts-1-hd',
    voice='alloy',
    input='Hello world! This is TTs test.',
)

response.write_to_file('hello_world.mp3')

# 재생
import IPython.display as ipd

ipd.Audio('hello_world.mp3')

```

```python
import json

# json 파일 열기
with open('../data/images/image_quiz_eng.json', 'r', encoding='utf-8') as f:
    eng_dict = json.load(f)


eng_dict
```

```python
voices = ['alloy', 'ash', 'coral', 'echo', 'fable', 'onyx', 'nova', 'sage', 'shimmer']

for q in eng_dict:
    no = q['no']
    quiz = q['eng']
    quiz.replace('- (1)', '- One.\t')
    quiz.replace('- (2)', '- Two.\t')
    quiz.replace('- (3)', '- Three.\t')
    quiz.replace('- (4)', '- Four.\t')

    print(q, quiz)

    voice = voices[no % len(voices)]

    response = client.audio.speech.create(
        model='tts-1-hd',
        voice=voice,
        input=f'#{no}. {quiz}'
    )

    response.write_to_file(f'../data/audio/{no}.mp3')
```

```python
# 재생
import IPython.display as ipd

ipd.Audio(f'../data/audio/1.mp3')
```
