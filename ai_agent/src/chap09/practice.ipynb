요구사항 1
1. 사용자의 질문에 대해 답변을 생성하는 체인을 작성하세요.
2. 체인은 아래와 같이 동작해야 합니다:
    * 입력 데이터: 사용자의 질문 (예: "인셉션 감독이 누구인가요?")
    * 출력 데이터: 질문에 대한 답변 answer (예 : "영화 "인셉션"의 감독은 크리스토퍼 놀란(Christopher Nolan)입니다.")
3. 답변을 생성하기 위해 다음 구성 요소를 사용하세요:
    * PromptTemplate: 사용자의 질문을 기반으로 적절한 답변을 생성하는 템플릿을 작성하세요.
    * ChatOpenAI: OpenAI 모델로 답변을 생성하세요.
    * StrOutputParser: 답변을 문자열로 반환하세요.
4. 이 체인을 chain1으로 정의하고, invoke 메서드를 사용하여 답변을 answer 변수에 저장하세요.


```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
```

```python
question = "인셉션 감독이 누구인가요?"
```

```python
template1 = """
당신은 영화를 추천해주는 AI 챗봇입니다. User의 질문에 대해 답변하시오.
question : {question}
"""

prompt = PromptTemplate(
    template=template1
)

model = ChatOpenAI(model_name="gpt-5-nano")

chain = prompt | model | StrOutputParser()
answer = chain.invoke({'question' : question})

print(answer)
```
